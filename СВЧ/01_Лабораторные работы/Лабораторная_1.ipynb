## 3. Обработка результатов

При одинаковых показаниях детектора $|S_{mn}|^2\,p_{\mathrm{ген}}=p_0$, где $p$ — мощность в линейных единицах. Поэтому для $m\ne n$:

$$
S_{mn,\mathrm{дБ}}=20\lg|S_{mn}|=P_0-P_{\mathrm{ген}}=-L_{mn}.
$$

В методичке разность $P_{\mathrm{ген}}-P_0$ записана как положительное затухание. Ниже все элементы приведены к единому обозначению $20\lg|S_{mn}|$.

Диагональные элементы вычисляем по формуле (3):

$$
S_{nn,\mathrm{дБ}}=20\lg\left(\frac{10^{\Delta P/20}-1}{10^{\Delta P/20}+1}\right).
$$

Матрица **модулей в децибелах** (строка — выход, столбец — вход):

$$
\left[S_{mn,\mathrm{дБ}}\right]=
\begin{pmatrix}
-18.11 & -17.35 & -17.14 \\
-20.62 & -9.73 & -1.55 \\
-1.22 & -20.85 & -19.21
\end{pmatrix}.
$$

**Вывод:** наименьшие потери получены для направлений $1\to3$ (1,22 дБ) и $3\to2$ (1,55 дБ); обратные направления ослаблены на 17,14 и 20,85 дБ. Это подтверждает невзаимность устройства. Вход 2 согласован хуже остальных: $|S_{22}|\approx0{,}326$ против 0,124 и 0,110. Одно из четырёх плеч циркулятора закрыто внутренней нагрузкой, поэтому замкнутый цикл между тремя доступными портами не требуется.

## 4. Контрольные вопросы

### 1. Объяснить работу щелевого моста

Щель в общей узкой стенке связывает два прямоугольных волновода. В области связи возбуждаются моды $H_{10}$ и $H_{20}$ с разными фазовыми скоростями. Длину щели выбирают так, чтобы разность набега фаз составляла $\pi/2$. При подаче сигнала в плечо 1 мощность поровну делится между плечами 3 и 4 со сдвигом фаз $90^\circ$; в плече 2 поля компенсируются.

### 2. Описать работу ферритового циркулятора

Первый щелевой мост делит сигнал на две волны. Ферритовые элементы и фазосдвигающая секция задают их относительную фазу так, что второй мост складывает волны в нужном выходе и гасит в другом. При обратном распространении феррит изменяет фазу иначе, поэтому отражённая волна направляется в другое плечо. В данной установке одно плечо постоянно соединено с согласованной нагрузкой.

### 3. Какими параметрами характеризуется циркулятор?

Основные параметры: прямые потери, развязка между плечами, коэффициенты отражения (или КСВН), рабочая полоса частот и допустимая мощность. Полное описание на заданной частоте даёт комплексная матрица $S$: диагональные элементы описывают отражение, остальные — передачу.

### 4. Из чего складывается погрешность измерения недиагональных элементов?

Основные причины: погрешность установки мощности и аттенюатора, дрейф генератора, неточность восстановления показания $\alpha_0$, рассогласование детектора и нагрузок, потери и неповторяемость соединений. Сравнение при одинаковом показании устраняет необходимость знать характеристику детектора. Работа на нижней границе полосы вольтметра может ухудшать точность сравнения.

### 5. В чём заключена погрешность измерения диагональных элементов?

Помимо ошибок мощности и отсчёта, влияют неточное нахождение максимума и минимума стоячей волны, возмущение поля зондом, потери измерительной линии и рассогласование нагрузок. При слабом отражении максимум и минимум близки, поэтому относительная погрешность $|S_{nn}|$ особенно велика.

### 6. Какой физический эффект используется в ферритовых циркуляторах?

Используется **гиротропия намагниченного феррита**: из-за прецессии магнитных моментов волны с правым и левым вращением магнитного поля имеют разные эффективные магнитные проницаемости. В данной конструкции при изменении направления распространения меняется вращение поля у ферритовой пластины, а значит, и фазовая скорость. Это создаёт невзаимный фазовый сдвиг.

### 7. Обосновать формулу (3)

Для $\rho=|S_{nn}|$ коэффициент стоячей волны равен

$$
K=\frac{E_{\max}}{E_{\min}}=\frac{1+\rho}{1-\rho},
\qquad \rho=\frac{K-1}{K+1}.
$$

Чтобы при переносе зонда из минимума в максимум сохранить показание детектора, мощность уменьшают в $K^2$ раз. Поэтому

$$
\Delta P=P_2^*-P_1^*=10\lg K^2=20\lg K,
\qquad K=10^{\Delta P/20}.
$$

Подставляя $K$ в выражение для $\rho$ и переходя к децибелам, получаем

$$
S_{nn,\mathrm{дБ}}=20\lg\left(\frac{10^{(P_2^*-P_1^*)/20}-1}{10^{(P_2^*-P_1^*)/20}+1}\right),
$$

что совпадает с формулой (3).

### 8. Как измерить фазы коэффициентов матрицы рассеяния?

Фазу отражения можно найти по смещению минимума стоячей волны относительно измерения с короткозамыкателем в той же опорной плоскости, зная длину волны в волноводе. Смещение на $\Delta x$ соответствует изменению фазы по модулю $2\beta\Delta x$, где $\beta=2\pi/\lambda_{\mathrm{в}}$; знак зависит от направления отсчёта координаты.

Для фазы передачи нужен опорный сигнал от генератора и фазовое сравнение с выходным сигналом, например фазометром или векторным анализатором цепей с калибровкой до плоскостей портов. По одним уровням мощности фазы определить нельзя.
