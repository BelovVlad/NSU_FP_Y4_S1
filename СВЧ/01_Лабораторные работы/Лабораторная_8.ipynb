# Лабораторная работа № 8
## Поля в объёмном резонаторе. Метод малого возмущающего тела

**Цель:** измерить резонансную частоту цилиндрического резонатора на моде $H_{011}$ и исследовать распределения квадратов электрического и магнитного полей с помощью диэлектрического и металлического шариков.

## 2. Метод обработки

Обозначим $\Delta f=f-f_0$, где $f_0$ — частота без шарика, $f$ — с шариком. При линейном качании

$$
f(t)=f_{\mathrm{start}}+\frac{t}{T}(f_{\mathrm{stop}}-f_{\mathrm{start}}),
\qquad \Delta f=0{,}5(t-t_0)\quad[\text{МГц}].
$$

Объём шарика $\Delta V=\pi d^3/6=1{,}13097\cdot10^{-7}$ м³. Введём квадраты полей, нормированные на запасённую энергию $W_0$:

$$
\mathcal E=\frac{\varepsilon_0|E|^2}{2W_0},\qquad
\mathcal H=\frac{\mu_0|H|^2}{2W_0}.
$$

Они имеют единицы м⁻³. Диэлектрический шарик определяет электрическое поле; металлический чувствителен к обоим полям:

$$
\boxed{\mathcal E=-\frac{\Delta f_d}{f_0 C_\varepsilon\Delta V}},
\qquad C_\varepsilon=3\frac{\varepsilon_r-1}{\varepsilon_r+2}=1{,}2,
$$

$$
\boxed{\mathcal H=2\mathcal E+\frac{2\Delta f_m}{3f_0\Delta V}}.
$$

Частоты в каждой дроби подставляются в одинаковых единицах. Для восстановления $\mathcal H$ значение $\mathcal E$ берётся в той же точке, при необходимости — линейной интерполяцией. За пределами измеренной диэлектрической серии магнитное поле не восстанавливается.

Показания шкал $x$ переводятся в координаты по исходной таблице:

$$
r_d=|223{,}2-x_d|,\quad r_m=|86{,}2-x_m|,\quad
z_d=98-x_d,\quad z_m=x_m-140{,}5\quad[\text{мм}].
$$

В графики включены исходные измерения; зеркально достроенные точки таблицы не использованы. Каждая кривая на графиках нормирована на собственный максимум, поэтому сравнивается её форма.

## 3. Теоретические зависимости

Для моды $H_{011}$ ненулевые компоненты — $E_\varphi$, $H_r$, $H_z$. При $\theta=3{,}832$, $C_H=\pi a/(h\theta)=0{,}5038$:

$$
|E|^2\propto J_1^2(\theta r/a)\sin^2(\pi z/h),
$$

$$
|H|^2\propto C_H^2J_1^2(\theta r/a)\cos^2(\pi z/h)
+J_0^2(\theta r/a)\sin^2(\pi z/h).
$$

$J_n$ — функции Бесселя. В магнитной зависимости использована формула методички с $\cos^2$ при $J_1^2$ и $\sin^2$ при $J_0^2$.

Резонансная частота:

$$
f_{0,\mathrm{теор}}=\frac{c}{2\pi}
\sqrt{\left(\frac{\theta}{a}\right)^2+\left(\frac{\pi}{h}\right)^2}
=2{,}686754\ \text{ГГц}.
$$

Из измеренного $t_{0r}=1{,}92$ с:

$$
f_{0,\mathrm{эксп}}=2{,}687+\frac{1{,}92}{4}(2{,}689-2{,}687)
=2{,}687960\ \text{ГГц}.
$$

Отличие от расчёта — 1,206 МГц, или **0,0449 %**. Теоретический максимум электрического поля находится при $r=36{,}61$ мм и $z=h/2=62$ мм.

## 5. Вывод

Получено $f_0=2{,}687960$ ГГц; отличие от расчёта для $H_{011}$ составляет 0,0449 %. Электрическое поле имеет кольцевой максимум при $r\approx36$–41 мм и максимум по высоте при $z=63$ мм. Поперечное магнитное поле максимально около оси. Эти особенности соответствуют структуре моды $H_{011}$.

Количественное согласие неполное: поперечные частотные сдвиги примерно вдвое меньше расчётных, а в продольной металлической серии знак сдвига расходится с теорией для $r=19$ мм. Требуется проверка опорного времени и траектории шарика. Основные источники погрешности — частотно-временная калибровка, отсчёт координат, конечный размер шарика и интерполяция электрического поля при восстановлении магнитного. Исходных данных для численной оценки погрешности нет.

## 6. Контрольные вопросы

**1. Как измеряют распределения полей?** Сначала находят $f_0$, затем измеряют частотные сдвиги при перемещении диэлектрического и металлического шариков. Из первой серии восстанавливают $\mathcal E$, из второй с учётом электрического вклада — $\mathcal H$.

**2. Как разделить компоненты полей?** Используют ориентированные пробные тела с известной анизотропной поляризуемостью или направленные зонды. Сфера измеряет сумму квадратов компонент; для относительных фаз нужны фазочувствительные измерения.

**3. Какие бывают простейшие колебания?** Электрические $E_{nim}$ (TM, $H_z=0$) и магнитные $H_{nim}$ (TE, $E_z=0$). У $E_{010}$ электрическое поле продольное и максимально на оси. У $H_{011}$ электрические линии — окружности вокруг оси, магнитное поле имеет радиальную и продольную компоненты.

**4. Собственные частоты цилиндрического резонатора:**

$$
f_{nim}=\frac{c}{2\pi\sqrt{\varepsilon_r\mu_r}}
\sqrt{\left(\frac{q_{ni}}a\right)^2+\left(\frac{m\pi}h\right)^2}.
$$

Для H-мод $q_{ni}$ — ненулевые корни $J'_n(q)=0$, $m\ge1$; для E-мод — корни $J_n(q)=0$, допускается $m=0$.

**5. Основные параметры резонатора:** собственные частоты и моды, структура поля, запасённая энергия, добротность $Q=\omega_0W_0/P_{\mathrm{пот}}$, полоса резонанса и связь с внешними линиями. Ширина резонанса в исходных данных не записана, поэтому $Q$ не вычислялась.

**6. Как определить характеристическое сопротивление?** По возмущению восстанавливают $E/\sqrt{W_0}$ вдоль выбранного зазора, интегрированием находят $U/\sqrt{W_0}$ и вычисляют $\rho=|U|^2/(2\omega_0W_0)$. Путь определения напряжения должен быть задан; у $H_{011}$ интеграл $E_z$ по оси равен нулю.

**7. Как найти проницаемость цилиндрического пробного тела на $E_{010}$?** Тонкий длинный диэлектрический цилиндр помещают на оси вдоль электрического поля. При малой продольной деполяризации

$$
\frac{\Delta f}{f_0}\simeq-\frac{\varepsilon_r-1}{2}\Delta V\,\mathcal E(0),
\qquad
\varepsilon_r\simeq1-\frac{2\Delta f}{f_0\Delta V\mathcal E(0)}.
$$

Для короткого цилиндра нужна поправка на его поляризуемость; сферическую формулу напрямую применять нельзя.
