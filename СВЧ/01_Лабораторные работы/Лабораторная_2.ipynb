## 2. Краткая теория и метод обработки

При одинаковой мощности возбуждения обозначим через $P$ мощность прямой волны в основном канале, через $P'$ — мощность полезной ответвлённой волны, через $P''$ — мощность нежелательного ответвления для обратного направления. Основные параметры:

$$
\xi=10\lg\frac{P}{P'},\qquad
\nu=10\lg\frac{P'}{P''}.
$$

Переходное ослабление $\xi$ определяет долю ответвляемой мощности, а направленность $\nu$ — способность различать прямую и обратную волны. Большая направленность соответствует лучшему разделению направлений.

Аттенюатор с ослаблением $m$ уменьшает мощность в $10^{m/10}$ раз. При постоянной мощности генератора равенство показаний одного и того же детектора означает равенство поступающих на него мощностей; знать его характеристику при этом не требуется, если она стабильна и однозначна в рабочей области.

Для измерения $\xi$ сравниваем передачу $A\to B$ при $n_B=50$ дБ и ответвление $A\to C$ при $n_C$:

$$
T\,10^{-n_B/10}=k\,10^{-n_C/10},\qquad
10\lg\frac{T}{k}=n_B-n_C,
$$

где $T=P_B/P_A$, $k=P_C/P_A$. При малых потерях и слабой связи $T\approx1$, поэтому, как в методичке,

$$
\boxed{\xi\approx n_B-n_C=50-n_C.}
$$

Строго метод сравнивает мощность в проходном порте с ответвлённой. Если учитывать потери основного канала $L_{AB}=-10\lg T$, ослабление относительно входной мощности равно $(n_B-n_C)+L_{AB}$. Значение $L_{AB}$ отдельно не измерялось; далее используется приближение методички.

Для направленности сравниваем слабый сигнал $B\to C$ при $m_B=0$ дБ и сильный сигнал $A\to C$ при $m_A$:

$$
k_{\mathrm{обр}}10^{-m_B/10}=k_{\mathrm{пр}}10^{-m_A/10},\qquad
\boxed{\nu=m_A-m_B=m_A.}
$$

Внутри каждой пары сравнений мощность и частота генератора, чувствительность индикатора и положение детектора должны сохраняться. Между разными частотами значение $\alpha_0$ может отличаться.

$$
u_\xi=\sqrt{u_{n_B}^2+u_{n_C}^2},\qquad
u_\nu=\sqrt{u_{m_A}^2+u_{m_B}^2}.

Для коррелированных отсчётов одного прибора следует учитывать ковариацию: $u^2(x-y)=u_x^2+u_y^2-2\operatorname{cov}(x,y)$. ## 6. Контрольные вопросы

### 1. Какими параметрами характеризуется направленный ответвитель?

Основные параметры — **переходное ослабление** $\xi=10\lg(P/P')$, **направленность** $\nu=10\lg(P'/P'')$, **рабочая полоса частот** и **коэффициенты отражения плеч** (либо КСВН). Также указывают потери проходного канала и допустимую мощность, ограниченную электрической прочностью волновода и нагрузкой побочного канала.

Развязка относительно входной мощности равна $I=10\lg(P/P'')=\xi+\nu$. Её следует отличать от направленности, в которой сравниваются два ответвлённых сигнала.

### 2. Объяснить работу направленного двухдырочного четвертьволнового ответвителя со связью на узкой стенке

Два одинаковых отверстия в общей узкой стенке связывают одинаковые прямоугольные волноводы. Для основной моды $H_{10}$ у узкой стенки имеется продольная составляющая магнитного поля; малые отверстия обеспечивают преимущественно магнитную связь.

Каждое отверстие возбуждает в побочном волноводе волны в обе стороны. В направлении прямой волны два вклада приходят к выходу с одинаковым суммарным набегом фазы: путь до второго отверстия в основном волноводе компенсируется меньшим путём от него до выхода в побочном. Поэтому они складываются.

В противоположном направлении разность путей равна $2d$, где $d$ — расстояние между отверстиями. При $d=\lambda_{\mathrm{в}}/4$ разность фаз составляет

$$
\Delta\varphi=2\beta d=\pi,\qquad \beta=\frac{2\pi}{\lambda_{\mathrm{в}}},
$$

и равные по амплитуде вклады взаимно гасятся. Здесь $\lambda_{\mathrm{в}}$ — **длина волны в волноводе**. При смене направления основной волны полезное ответвление идёт к противоположному концу побочного волновода и поглощается встроенной нагрузкой. Это взаимное пассивное устройство; направленность не означает невзаимности.

### 3. Описать методику проведения измерений параметров направленного ответвителя

Используется метод замещения с аттенюатором, включённым между генератором и ответвителем.

1. **Направленность:** генератор в $B$, индикатор в $C$, нагрузка в $A$. При $m_B=0$ дБ записывают $\alpha_0$. Затем меняют местами генератор и нагрузку и при $A\to C$ увеличивают ослабление до $m_A$, восстанавливая $\alpha_0$. Получают $\nu=m_A-m_B$.
2. **Переходное ослабление:** при $A\to B$ и $n_B=50$ дБ записывают $\alpha_0$. Затем индикатор из $B$ переносят в $C$, нагрузку — из $C$ в $B$, и уменьшают ослабление до $n_C$, восстанавливая $\alpha_0$. Получают $\xi\approx n_B-n_C$.
3. Повторяют сравнения на каждой частоте и строят $\xi(f)$ и $\nu(f)$.

Равенство показаний позволяет обойтись без калибровки зависимости сигнала детектора от мощности. Квадратичность детектора для этого метода не обязательна, но необходимы стабильность его характеристики, одинаковый диапазон индикатора и неизменность генератора в пределах пары сравнений.

### 4. Посмотреть в основной канал исследуемого ответвителя и объяснить структурные особенности элементов связи

По методичке связь выполнена в общей **узкой стенке** основного и побочного волноводов. Отверстия или щели пропускают часть поля в побочный канал. Их размеры задают силу связи, а продольные расстояния — относительные фазы ответвлённых волн. В простейшей двухдырочной конструкции одинаковые элементы размещают через $\lambda_{\mathrm{в}}/4$ на расчётной частоте.

В многоэлементной конструкции изменение размеров отверстий позволяет задать веса отдельных вкладов и обеспечить подавление нежелательной волны в более широкой полосе. Если при осмотре видны отверстия разного размера, это может быть частью такого распределения связи. Встроенная согласованная нагрузка поглощает волну, уходящую к закрытому концу побочного канала, и уменьшает её обратное отражение.

**Ограничение исходных данных:** фотография внутренней стенки и запись её осмотра отсутствуют. Поэтому число, форма, размеры и фактическое распределение отверстий данного экземпляра здесь не утверждаются; приведено объяснение их назначения.

### 5. Как меняется направленность двухдырочного ответвителя при изменении расстояния между отверстиями от 0 до $\lambda$?

Рассмотрим одинаковые слабо связанные отверстия и одинаковые волноводы при постоянной частоте. Под $\lambda$ понимаем $\lambda_{\mathrm{в}}$. Если вклад одного отверстия имеет амплитуду $a$, то, с точностью до общей фазы,

$$
A_{\mathrm{пр}}=2a,\qquad
A_{\mathrm{обр}}=a(1+e^{-2i\beta d})
=2a e^{-i\beta d}\cos(\beta d).
$$

Следовательно,

$$
\boxed{\nu(d)=-20\lg\left|\cos\frac{2\pi d}{\lambda_{\mathrm{в}}}\right|.}
$$

| $d/\lambda_{\mathrm{в}}$ | 0 | $1/4$ | $1/2$ | $3/4$ | 1 |
| :--- | :---: | :---: | :---: | :---: | :---: |
| $\nu$ в идеальной модели | 0 дБ | $\infty$ | 0 дБ | $\infty$ | 0 дБ |

Таким образом, направленность дважды возрастает до максимума и падает до нуля; период равен $\lambda_{\mathrm{в}}/2$. В реальном приборе максимумы конечны из-за неравенства вкладов, погрешностей геометрии и отражений. Предел $d\to0$ здесь формальный: при сближении конечных отверстий меняются их взаимное влияние и геометрия.

### 6. Как меняется переходное ослабление при таком же изменении расстояния?

В той же модели слабой связи вклады в прямом направлении складываются синфазно при любом $d$, поскольку фазовые постоянные обоих волноводов одинаковы. Поэтому

$$
P'\propto|2a|^2=4|a|^2,\qquad
\xi(d)\approx\mathrm{const}.
$$

По сравнению с одним таким отверстием полезная ответвлённая мощность в идеальном приближении больше в четыре раза, а переходное ослабление меньше на $10\lg4\approx6{,}02$ дБ.

Это утверждение относится к **изменению расстояния при фиксированной частоте**, одинаковой силе связи отверстий и пренебрежимо малом истощении основной волны. При изменении частоты меняются свойства связи и длина волны; поэтому измеренная $\xi(f)$ может изменяться. Взаимодействие близких отверстий и неодинаковые волноводы также нарушают независимость от $d$.

### 7. Какова погрешность измерения коэффициента отражения при конечной направленности?

Пусть ответвитель ориентирован на отражённую волну, его полезный коэффициент передачи учтён при калибровке, а остальные источники ошибки пока отсутствуют. Помимо сигнала $\Gamma$ в измерительный канал проникает часть падающей волны с нормированной амплитудой

$$
\varepsilon=10^{-\nu/20}.
$$

Эти сигналы когерентны, поэтому складываются **амплитуды**, а не мощности:

$$
\rho_{\mathrm{изм}}=|\Gamma+\varepsilon e^{i\varphi}|,\qquad
\rho=|\Gamma|,
$$

$$
\rho_{\mathrm{изм}}^2=\rho^2+\varepsilon^2+2\rho\varepsilon\cos\theta.
$$

Неизвестная относительная фаза $\theta$ определяет знак ошибки. По неравенству треугольника

$$
|\rho-\varepsilon|\le\rho_{\mathrm{изм}}\le\rho+\varepsilon,
\qquad
\boxed{|\Delta\rho|\le\varepsilon},\qquad
\frac{|\Delta\rho|}{\rho}\le\frac{10^{-\nu/20}}{\rho}\quad(\rho>0).
$$

Например, при 9,3 ГГц $\varepsilon\approx0{,}053$, а при 9,7 ГГц $\varepsilon\approx0{,}143$. Это абсолютные ошибки модуля, а не 5,3 % и 14,3 % от измеряемого значения. При истинном $\rho=0{,}1$ предельные относительные ошибки в этой модели составляют около 53 % и 143 %. Для идеально согласованной нагрузки прибор покажет паразитное отражение порядка $\varepsilon$.

**При нормировке на короткое замыкание по рис. 2** паразитный сигнал присутствует и в эталонном измерении. В приближении неизменной падающей волны:

$$
\rho_{\mathrm{изм}}=
\frac{|\Gamma+\delta|}{|\Gamma_{\mathrm{КЗ}}+\delta|},
\qquad |\delta|=\varepsilon,\quad |\Gamma_{\mathrm{КЗ}}|=1.
$$

Из $1-\varepsilon\le|\Gamma_{\mathrm{КЗ}}+\delta|\le1+\varepsilon$ следует консервативная оценка

$$
|\Delta\rho|\le\frac{\varepsilon(1+\rho)}{1-\varepsilon},\qquad \varepsilon<1.
$$

Таким образом, $10^{-\nu/20}$ — характерный масштаб ошибки от направленности, а точная граница зависит также от способа калибровки.

### 8. Из чего складывается погрешность измерения $\nu$ и $\xi$ по предложенной методике?

Основные составляющие:

- погрешность калибровки и отсчёта ослабления аттенюатора, его частотная зависимость и повторяемость установки;
- неточное восстановление $\alpha_0$, шумы и дрейф детектора и вольтметра; особенно существенны при слабом сигнале обратного ответвления;
- изменение мощности, частоты или глубины модуляции генератора между двумя сравнениями;
- рассогласование генератора, детектора и нагрузок, многократные отражения и изменение входного согласования при перестановке портов;
- потери, загрязнение и неповторяемость фланцевых соединений; для $\xi$ также приближение равенства входной и проходной мощности.

Метод одинакового показания устраняет необходимость знать абсолютную чувствительность и нелинейную характеристику детектора. Однако он не устраняет изменение этой характеристики во времени, ошибку установки мощности и влияние отражений. Для численной оценки нужны характеристики приборов и повторные измерения.

### 9. Можно ли по схеме рис. 2 измерить направленность ответвителя?

**Да, если дополнить схему сравнением с согласованной нагрузкой.** Ответвитель на рис. 2 ориентирован на отражённую волну. При согласованной нагрузке истинное отражение практически отсутствует и показание $\alpha_{\mathrm{согл}}$ определяется главным образом паразитным проникновением падающей волны. После установки короткозамыкателя отражение по модулю равно единице; при высокой направленности показание $\alpha_{\mathrm{КЗ}}$ преимущественно определяется полезной связью.

При одинаковой падающей мощности, квадратичной характеристике детектора и отсутствии значимого фонового сигнала:

$$
\boxed{\nu\approx10\lg\frac{\alpha_{\mathrm{КЗ}}}{\alpha_{\mathrm{согл}}}.}
$$

Это приближённая формула: остаточное отражение согласованной нагрузки и интерференция полезного сигнала с паразитным при КЗ вызывают ошибку. Одного показания с неподвижным короткозамыкателем недостаточно. Генератор развязывают аттенюатором или ферритовым вентилем, чтобы смена нагрузки не меняла падающую волну.

Если имеется **подвижный короткозамыкатель**, можно менять фазу отражения и определить максимальное и минимальное показания на протяжении полного фазового периода (смещение КЗ на $\lambda_{\mathrm{в}}/2$). При неизменных амплитудах и полезном сигнале больше паразитного:

$$
\alpha_{\max}=K(A+B)^2,\qquad
\alpha_{\min}=K(A-B)^2,
$$

$$
\nu=20\lg\frac{A}{B}
=20\lg\frac{\sqrt{\alpha_{\max}}+\sqrt{\alpha_{\min}}}
{\sqrt{\alpha_{\max}}-\sqrt{\alpha_{\min}}}.
$$

Здесь $A$ — амплитуда полезного ответвления от полностью отражённой волны, $B$ — амплитуда паразитного сигнала, $K$ — коэффициент квадратичного детектора. Этот вариант требует возможности перемещать КЗ и стабильной мощности; наличие такого короткозамыкателя в исходной записи не указано.
