# Лабораторная работа № 5
## Измерение параметров взаимных четырёхполюсников без потерь

**Цель:** проверить квадратичность детектора измерительной линии, измерить длины волн в коаксиальном и волноводном трактах, определить КСВН и модули коэффициентов отражения и передачи волноводно-коаксиального перехода методами Татаринова и Вайсфлоха.

## 2. Краткая теория и метод обработки

Матрица рассеяния связывает нормированные на мощность падающие и выходящие волны:

$$
\begin{pmatrix}b_1\\b_2\end{pmatrix}
=S\begin{pmatrix}a_1\\a_2\end{pmatrix},\qquad
S=\begin{pmatrix}S_{11}&S_{12}\\S_{21}&S_{22}\end{pmatrix}.
$$

Взаимность означает $S_{12}=S_{21}$, отсутствие потерь — $S^\dagger S=I$. При согласованной нагрузке второго порта коэффициент отражения первого порта равен $S_{11}$, а коэффициент передачи — $S_{21}$. Поэтому

$$
\rho=|\Gamma|=|S_{11}|,\qquad |T|=|S_{21}|=\sqrt{1-\rho^2}.
$$

Для стоячей волны $E_{\max}=|E_+|(1+\rho)$ и $E_{\min}=|E_+|(1-\rho)$. Коэффициент стоячей волны напряжения:

$$
K=\frac{1+\rho}{1-\rho},\qquad
\boxed{\rho=\frac{K-1}{K+1}.}
$$

### Метод Татаринова

Второй порт нагружают на согласованную нагрузку. В измерительной линии находят максимум и минимум поля. Для квадратичного детектора $\alpha\propto|E|^2$, поэтому

$$
\boxed{K_{\mathrm{Т}}=\sqrt{\frac{\alpha_{\max}}{\alpha_{\min}}}.}
$$

### Метод Вайсфлоха

Вместо согласованной нагрузки устанавливают подвижный короткозамыкающий поршень. Для каждого положения $P$ находят положение минимума $L$ в измерительной линии:

$$
x_2=P-P_0,\qquad x_1=L-L_0,\qquad
\beta(x_2)=\frac{x_1}{\Lambda_1}-\frac{x_2}{\Lambda_2}.
$$

Величина $\beta$ безразмерна. По рис. 9 методички обозначение $M$ относится к **полному размаху** между максимумом и минимумом:

$$
M=\beta_{\max}-\beta_{\min},\qquad
A=\frac{M}{2},\qquad
\beta_c=\frac{\beta_{\max}+\beta_{\min}}{2}.
$$

Здесь $A$ — амплитуда относительно средней линии. Формулы обработки:

$$
\boxed{K_{\mathrm{В}}=\frac{1+\sin(\pi M)}{1-\sin(\pi M)},
\qquad \rho_{\mathrm{В}}=\sin(\pi M)=\sin(2\pi A).}
$$

Изменение начала отсчёта сдвигает кривую по вертикали и не меняет её размах. Метод использует положения минимумов, поэтому квадратичность детектора для него не обязательна.

### Анализ расхождения методов

Модули отражения различаются на $0{,}05246$; результат Татаринова примерно на **46,7 %** больше результата Вайсфлоха. Изменение определения $M$ уменьшает исходное расхождение, но не устраняет его.

Метод Татаринова чувствителен к остаточному отражению согласованной нагрузки и характеристике детектора. Метод Вайсфлоха чувствителен к нахождению минимумов, координатам поршня и зонда, значениям обеих длин волн и дискретности снятия кривой. Смена соединений и дрейф частоты также могут влиять на сравнение. Двухмиллиметровый шаг не гарантирует попадания в истинные экстремумы; при остальных идеальных условиях размах по точкам может быть занижен.

При слабых неоднородностях метод Вайсфлоха обычно предпочтительнее, поскольку короткозамыкатель легче приблизить к идеальному элементу, чем согласованную нагрузку. Однако без повторных измерений и характеристик приборов нельзя численно установить неопределённость или доказать согласие результатов в её пределах. Передаваемая мощность здесь вычислена в предположении отсутствия потерь, а не измерена независимо.

Для независимых неопределённостей показаний детектора:

$$
\frac{u_K}{K}=\frac12\sqrt{\left(\frac{u_{\alpha_{\max}}}{\alpha_{\max}}\right)^2+
\left(\frac{u_{\alpha_{\min}}}{\alpha_{\min}}\right)^2},\qquad
u_\rho=\frac{2u_K}{(K+1)^2}.
$$

Для метода Вайсфлоха $u_\rho\approx\pi|\cos(\pi M)|u_M$. В $u_M$ входят ошибки координат и длин волн; общее значение $\Lambda_1$ или $\Lambda_2$ создаёт корреляцию ошибок разных точек. Численная оценка по имеющимся данным не задана.

Во второй серии по методу Татаринова получены $K=1{,}4530$, $|\Gamma|=0{,}18466$, $|T|=0{,}98280$. Предварительная оценка по методу Вайсфлоха: $K\approx1{,}5333$, $|\Gamma|\approx0{,}2105$, $|T|\approx0{,}9776$; она требует уточнения длины волны и проверки координат.

Результаты указывают на небольшое отражение: в модели без потерь доля переданной мощности составляет около **95,6–98,7 %**, включая предварительную оценку второй серии. Различия методов могут быть связаны с несовершенством нагрузок, ошибками нахождения минимумов и конечным шагом снятия S-кривой. Без оценок неопределённости статистическое согласие результатов не установлено. Фазы коэффициентов не измерялись, поэтому полную комплексную матрицу рассеяния восстановить нельзя.

## 6. Контрольные вопросы

### 1. Какой основной тип волн распространяется в коаксиальной линии?

Основная волна — **TEM**: $E_z=H_z=0$, обе составляющие поля поперечны оси линии. Электрическое поле направлено преимущественно радиально, магнитное — по окружностям вокруг внутреннего проводника. Для однородного диэлектрика $v=1/\sqrt{\mu\varepsilon}$; в воздушной линии $v\approx c$. У TEM-волны нет ненулевой частоты отсечки, но на достаточно высоких частотах могут возбуждаться высшие типы волн.

### 2. В чём заключается суть методов Татаринова и Вайсфлоха?

В методе Татаринова второй порт нагружают согласованно и измеряют отношение максимального и минимального поля на первом порте. Для квадратичного детектора $K=\sqrt{\alpha_{\max}/\alpha_{\min}}$.

В методе Вайсфлоха второй порт замыкают подвижным поршнем, отслеживают перемещение минимума на первом порте и строят $\beta=x_1/\Lambda_1-x_2/\Lambda_2$. По полному размаху $M$ определяют $|\Gamma|=\sin(\pi M)$ и $K=(1+|\Gamma|)/(1-|\Gamma|)$. При коротком замыкании отражение всего нагруженного тракта по модулю близко к единице; искомое собственное отражение перехода извлекают из изменения положения узла, а не из КСВН этого замкнутого тракта.

### 3. Какими параметрами характеризуются четырёхполюсники?

Полное линейное описание на заданной частоте даётся комплексной матрицей $S$; возможны также матрицы сопротивлений $Z$, проводимостей $Y$ и передачи $ABCD$. $S_{11}$ и $S_{22}$ описывают отражение от портов, $S_{21}$ и $S_{12}$ — передачу. Практические параметры: КСВН, потери передачи, возвратные потери, фазы и частотные зависимости. Определение каждого $S_{ij}$ предполагает отсутствие падающих волн на остальных портах, то есть их согласованную нагрузку.

### 4. Написать матрицу рассеяния взаимного четырёхполюсника без потерь

При нормировке волн на мощность выполняются

$$
S_{12}=S_{21},\qquad S^\dagger S=I,
$$

$$
|S_{11}|^2+|S_{21}|^2=1,\qquad
|S_{12}|^2+|S_{22}|^2=1,\qquad
S_{11}^*S_{12}+S_{21}^*S_{22}=0.
$$

Отсюда $|S_{11}|=|S_{22}|$. Взаимность сама по себе **не означает** $S_{11}=S_{22}$: это дополнительное свойство симметричного устройства или подходящего выбора плоскостей отсчёта. Если $S_{11}=S_{22}=\rho e^{i\varphi}$, то взаимная фаза отражения и передачи равна $\pm\pi/2$ по модулю $2\pi$. Один допустимый выбор:

$$
S=e^{i\varphi}
\begin{pmatrix}
\rho & -i\sqrt{1-\rho^2}\\
-i\sqrt{1-\rho^2}&\rho
\end{pmatrix}.
$$

Для измеренных значений $\rho$ эта запись описывает семейство матриц: фаза $\varphi$ в работе не определена. Формула для передачи относится к нормированным волнам; при разных сопротивлениях портов её нельзя без пересчёта считать отношением физических напряжений.

### 5. Какое расстояние в длинах волн между точками $x_2'$ и $x_2''$ на рис. 9?

На рис. 9 отмечены соседние пересечения средней линии с противоположными наклонами. В симметричной идеальной модели расстояние между ними равно **$\Lambda_2/4$**. Полный период кривой и расстояние между пересечениями с одинаковым наклоном — **$\Lambda_2/2$**, поскольку при таком смещении поршня фаза отражённой волны меняется на $2\pi$.

Для $\Lambda_2=42{,}4$ мм эти расстояния составляют 10,6 и 21,2 мм. Это теоретические значения; дискретные измеренные точки не обязаны попадать точно в пересечения.

### 6. Указать источники погрешности измерения коэффициента отражения

Общие причины: дрейф частоты и мощности генератора, нестабильность соединений, потери линии, возмущение поля зондом, изменение его глубины и чувствительности, ошибки отсчёта координат и неполное соблюдение модели без потерь.

Для метода Татаринова особенно существенны остаточное отражение согласованной нагрузки, нелинейность детектора, фон индикатора и ошибки нахождения максимума и минимума. Сигналы от перехода и нагрузки складываются когерентно, поэтому остаточное отражение нагрузки нельзя просто вычесть по мощности.

Для метода Вайсфлоха существенны неидеальность короткозамыкателя, ошибки координат $P$ и $L$, неправильное сопровождение одного и того же узла, ошибки $\Lambda_1$, $\Lambda_2$ и пропуск экстремумов из-за конечного шага. Неверная замена полного размаха полуразмахом — ошибка обработки, а не случайная погрешность измерений.

### 7. Что такое эквивалентная ширина щели?

В плоскостной конструкции коаксиальной измерительной линии поле быстро убывает к боковым краям. Её открытая щель может быть физически заметно шире, чем щель обычной концентрической линии, сохраняя малое возмущение поля. **Эквивалентная ширина** — угловая ширина щели в соответствующей концентрической коаксиальной конфигурации, полученной конформным преобразованием поля.

В обозначениях рис. 2 методички:

$$
W=\frac{4}{\sinh(\pi d/(2A))}\quad\text{рад},
$$

где $A$ и $d$ — геометрические размеры этой конструкции, а $2A$ — ширина щели плоскостной линии. При большом отношении $d/A$ величина $W$ мала, поэтому излучение через щель и изменение волнового сопротивления незначительны. Равенство конфигураций относится к структуре поля, а не к равенству физических размеров щелей.

### 8. Сравнить рабочую длину волны, измеренную тремя способами

По частоте генератора: $\lambda_0=c/f=31{,}4136$ мм. По коаксиальной линии: $\lambda_0\approx\Lambda_1=31{,}4$ мм. По поршню: сначала $\Lambda_2=42{,}4$ мм, затем с учётом дисперсии $H_{10}$ и $a=23$ мм получаем $\lambda_0=31{,}1764$ мм.

Отличия приведённых длин от значения генератора составляют около −0,043 % и −0,755 %. Нельзя интерпретировать разность $42{,}4-31{,}4$ мм как ошибку: это длины волн в разных типах линии. Без оценок неопределённости нельзя доказать статистическое согласие измерений.

## 7. Вывод формул для подготовки к защите



### Квадратичность детектора и расстояние между минимумами

При коротком замыкании две встречные волны равной амплитуды образуют стоячую волну. Если $l_0$ — положение узла, то

$$
E(l)=E_+\left[e^{-ik_1(l-l_0)}-e^{ik_1(l-l_0)}\right]
=-2iE_+\sin[k_1(l-l_0)],\qquad k_1=\frac{2\pi}{\Lambda_1}.
$$

Для квадратичного детектора $\alpha\propto|E|^2\propto\sin^2[k_1(l-l_0)]$. Нули синуса повторяются через $\pi/k_1=\Lambda_1/2$, что даёт формулу измерения длины волны по двум соседним минимумам.

### Связь КСВН с модулем отражения

В максимуме падающая и отражённая волны складываются в фазе, в минимуме — в противофазе:

$$
E_{\max}=|E_+|+|E_-|=|E_+|(1+\rho),\qquad
E_{\min}=|E_+|-|E_-|=|E_+|(1-\rho).
$$

Деление и решение относительно $\rho$ дают $K=(1+\rho)/(1-\rho)$ и $\rho=(K-1)/(K+1)$. Квадратичный закон детектора даёт $\alpha_{\max}/\alpha_{\min}=K^2$, поэтому в методе Татаринова нужен квадратный корень.

### Дисперсия волны $H_{10}$

В прямоугольном воздушном волноводе поперечное волновое число основной волны $k_\perp=\pi/a$. Из уравнения Гельмгольца

$$
k_0^2=k_z^2+k_\perp^2,\qquad
k_0=\frac{2\pi f}{c},\qquad k_z=\frac{2\pi}{\Lambda_2}.
$$

Разделив на $(2\pi)^2$, получаем

$$
\frac{f^2}{c^2}=\frac{1}{\Lambda_2^2}+\frac{1}{(2a)^2},\qquad
f=c\sqrt{\frac{1}{\Lambda_2^2}+\frac{1}{(2a)^2}}.
$$

Частота отсечки $f_c=c/(2a)\approx6{,}522$ ГГц. При $f>f_c$ волна распространяется, а $\Lambda_2=\lambda_0/\sqrt{1-(f_c/f)^2}>\lambda_0$. Различие длин волн вызвано разными модами и дисперсией, а не просто различием волновых сопротивлений.

### Почему в методе Вайсфлоха нужен полный размах

Сменой плоскостей отсчёта взаимный четырёхполюсник без потерь можно привести к форме $S_{11}=S_{22}=\rho$ и $S_{12}=S_{21}=i\sqrt{1-\rho^2}$. Это удобный выбор фаз для вывода; он не определяет экспериментальные фазы.

Для произвольной нагрузки с коэффициентом отражения $\Gamma_L$:

$$
\Gamma_{\mathrm{вх}}=S_{11}+\frac{S_{12}S_{21}\Gamma_L}{1-S_{22}\Gamma_L}.
$$

У идеального короткозамыкателя $|\Gamma_L|=1$. Запишем $\Gamma_L=e^{i\theta}$, включив постоянный знак короткого замыкания в $\theta$. Тогда

$$
\Gamma_{\mathrm{вх}}=\frac{\rho-e^{i\theta}}{1-\rho e^{i\theta}}
=-e^{i\theta}\frac{1-\rho e^{-i\theta}}{1-\rho e^{i\theta}}.
$$

Его фазовое отклонение от линейной зависимости при $\rho=0$ равно

$$
\delta=-2\arg(1-\rho e^{i\theta}).
$$

Комплексное число $1-\rho e^{i\theta}$ описывает окружность радиуса $\rho$ с центром в точке 1. Наибольший угол касательной из начала координат равен $\arcsin\rho$, поэтому $|\delta|_{\max}=2\arcsin\rho$ относительно середины диапазона.

Смещение минимума на долю длины волны меняет фазу отражения на $4\pi x/\Lambda$. Поэтому, с точностью до знака координат и постоянного сдвига $\beta_c$,

$$
|\beta-\beta_c|_{\max}=\frac{2\arcsin\rho}{4\pi}
=\frac{\arcsin\rho}{2\pi}=A.
$$

Таким образом,

$$
M=2A=\frac{\arcsin\rho}{\pi},\qquad
\boxed{\rho=\sin(\pi M)=\sin(2\pi A).}
$$

Подстановка в $K=(1+\rho)/(1-\rho)$ даёт формулу (4) методички. При малом отражении $\rho\approx\pi M$. Этот вывод также показывает, почему подстановка полуразмаха в $\sin(\pi M)$ почти вдвое занижает результат.
