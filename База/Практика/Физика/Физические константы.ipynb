# Физические константы

Выберите раздел или найдите величину по названию и обозначению: «Больцмана», `hbar`, «масса электрона».

Значения основаны на **CODATA 2022**. Десятичные записи экспериментальных и производных констант округлены; неопределённости приведены в таблице NIST в разделе «Источники». Значения $c$, $h$, $e$, $k_{\mathrm{B}}$ и $N_{\mathrm{A}}$ в СИ заданы точно.

В электрических строках СГС используется электростатический заряд, в магнитных — гауссовы единицы. Переводы энергии, приставки и размерности находятся в [справочнике систем единиц](Системы%20единиц%20и%20фундаментальные%20соотношения.ipynb).

### Магнетоны

| Величина | Обозначение | СИ | СГС |
| --- | --- | --- | --- |
| Магнетон Бора | $\mu_{\mathrm{B}}$ | $9.2740100657\times10^{-24}\,\dfrac{\text{Дж}}{\text{Тл}}$ | $9.2740100657\times10^{-21}\,\dfrac{\text{эрг}}{\text{Гс}}$ |
| Ядерный магнетон | $\mu_{\mathrm{N}}$ | $5.0507837393\times10^{-27}\,\dfrac{\text{Дж}}{\text{Тл}}$ | $5.0507837393\times10^{-24}\,\dfrac{\text{эрг}}{\text{Гс}}$ |


$$
\varepsilon_0\mu_0c^2=1,\qquad
\alpha=\frac{e^2}{4\pi\varepsilon_0\hbar c},\qquad
\mu_\mathrm{B}=\frac{e\hbar}{2m_e},\quad
\mu_\mathrm{N}=\frac{e\hbar}{2m_p}.
$$

С 2019 года $\mu_0$ и $\varepsilon_0$ определяются экспериментально: равенство $\mu_0=4\pi\times10^{-7}\,\dfrac{\text{Гн}}{\text{м}}$ больше не является точным. Для обычных учебных задач это остаётся очень хорошим приближением.

В гауссовой СГС $\varepsilon_0$ и $\mu_0$ не входят в уравнения в том же виде, что в СИ. Формула тонкой структуры там имеет вид $\alpha=e_\mathrm{G}^2/(\hbar c)$; магнитный момент удобно измерять в эрг/Гс: $1\,\dfrac{\text{Дж}}{\text{Тл}}$ $=10^3\,\dfrac{\text{эрг}}{\text{Гс}}$.

### Атомные длины и сечения

| Величина | Обозначение | СИ | СГС |
| --- | --- | --- | --- |
| Боровский радиус | $a_\infty=a_0$ | $5.29177210544\times10^{-11}\,\text{м}$ | $5.29177210544\times10^{-9}\,\text{см}$ |
| Классический радиус электрона | $r_e$ | $2.8179403205\times10^{-15}\,\text{м}$ | $2.8179403205\times10^{-13}\,\text{см}$ |
| Комптоновская длина волны электрона | $\lambda_{\mathrm{C}}=\dfrac{h}{m_ec}$ | $2.42631023538\times10^{-12}\,\text{м}$ | $2.42631023538\times10^{-10}\,\text{см}$ |
| Приведённая комптоновская длина | $\bar\lambda_{\mathrm{C}}=\dfrac{\hbar}{m_ec}$ | $3.8615926744\times10^{-13}\,\text{м}$ | $3.8615926744\times10^{-11}\,\text{см}$ |
| Атомное сечение (масштаб площади) | $\pi a_0^2$ | $8.7973554\times10^{-21}\,\text{м}^{2}$ | $8.7973554\times10^{-17}\,\text{см}^{2}$ |
| Сечение Томсона | $\sigma_{\mathrm{T}}=\dfrac{8\pi r_e^2}{3}$ | $6.6524587051\times10^{-29}\,\text{м}^{2}$ | $6.6524587051\times10^{-25}\,\text{см}^{2}$ |

$$
a_0=\frac{4\pi\varepsilon_0\hbar^2}{m_e e^2},\qquad
r_e=\frac{e^2}{4\pi\varepsilon_0m_ec^2},\qquad
\lambda_\mathrm{C}=2\pi\bar\lambda_\mathrm{C}.
$$

**$\pi a_0^2$ — масштаб площади**, а не универсальное сечение столкновения атомов: реальное сечение зависит от процесса и энергии.


Обозначение $a_\infty$ из исходной таблицы сохранено; здесь оно означает тот же боровский радиус $a_0$.


### Атомные энергии и постоянная Ридберга

| Величина | Обозначение | СИ | СГС |
| --- | --- | --- | --- |
| Постоянная Ридберга | $R_\infty$ | $1.0973731568157\times10^7\,\text{м}^{-1}$ | $1.0973731568157\times10^5\,\text{см}^{-1}$ |
| Энергия Ридберга | $E_{\mathrm{Ry}}=hcR_\infty$ | $2.1798723611030\times10^{-18}\,\text{Дж}$ | $2.1798723611030\times10^{-11}\,\text{эрг}$ |
| Энергия Хартри | $E_{\mathrm{h}}=2hcR_\infty$ | $4.3597447222060\times10^{-18}\,\text{Дж}$ | $4.3597447222060\times10^{-11}\,\text{эрг}$ |

$$
R_\infty=\frac{m_e e^4}{8\varepsilon_0^2 c h^3}
=\frac{\alpha^2m_ec}{2h},\qquad
E_\mathrm{Ry}=\frac{\alpha^2m_ec^2}{2}.
$$

Индекс $\infty$ означает бесконечно тяжёлое ядро. Для спектра водородоподобного атома в нерелятивистском приближении учитывают приведённую массу: $R_M=R_\infty\,m_\mathrm{red}/m_e$, где $m_\mathrm{red}=m_eM/(m_e+M)$.
