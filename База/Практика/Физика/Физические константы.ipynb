# Физические константы

Значения для расчётов, сгруппированные по смыслу. СИ и СГС стоят рядом: не нужно искать одну величину в двух длинных таблицах.

**Найти величину:** введите название или обозначение в поиск сверху, например «Больцмана», `hbar`, «масса электрона». Оглавление справа ведёт к разделам; на телефоне оно открывается кнопкой «Содержание».

> **Как читать числа.** Основа — CODATA 2022. Знак **†** отмечает величины с точным значением по определению СИ; многоточие означает показанную часть бесконечной десятичной записи. Остальные значения округлены для учебных расчётов. Для расчёта неопределённостей используйте полную таблицу NIST в разделе «Источники».

Переводы эВ ↔ Дж ↔ К, приставки и правила работы с СГС находятся в [справочнике систем единиц](Системы%20единиц%20и%20фундаментальные%20соотношения.ipynb).

### Магнетоны

| Величина | Обозначение | Значение | Единица |
| --- | --- | --- | --- |
| Магнетон Бора | $\mu_\mathrm{B}$ | $9.2740100657\times10^{-24}$ | Дж/Тл |
| Ядерный магнетон | $\mu_\mathrm{N}$ | $5.0507837393\times10^{-27}$ | Дж/Тл |


$$
\varepsilon_0\mu_0c^2=1,\qquad
\alpha=\frac{e^2}{4\pi\varepsilon_0\hbar c},\qquad
\mu_\mathrm{B}=\frac{e\hbar}{2m_e},\quad
\mu_\mathrm{N}=\frac{e\hbar}{2m_p}.
$$

С 2019 года $\mu_0$ и $\varepsilon_0$ определяются экспериментально: равенство $\mu_0=4\pi\times10^{-7}$ Гн/м больше не является точным. Для обычных учебных задач это остаётся очень хорошим приближением.

В гауссовой СГС $\varepsilon_0$ и $\mu_0$ не входят в уравнения в том же виде, что в СИ. Формула тонкой структуры там имеет вид $\alpha=e_\mathrm{G}^2/(\hbar c)$; магнитный момент удобно измерять в эрг/Гс: $1$ Дж/Тл $=10^3$ эрг/Гс.

### Атомные длины и сечения

| Величина | Обозначение | СИ | СГС |
| --- | --- | --- | --- |
| Боровский радиус | $a_0$ | $5.29177210544\times10^{-11}$ м | $5.29177210544\times10^{-9}$ см |
| Классический радиус электрона | $r_e$ | $2.8179403205\times10^{-15}$ м | $2.8179403205\times10^{-13}$ см |
| Комптоновская длина волны электрона | $\lambda_\mathrm{C}=h/(m_ec)$ | $2.42631023538\times10^{-12}$ м | $2.42631023538\times10^{-10}$ см |
| Приведённая комптоновская длина | $\bar\lambda_\mathrm{C}=\hbar/(m_ec)$ | $3.8615926744\times10^{-13}$ м | $3.8615926744\times10^{-11}$ см |
| Характерная атомная площадь | $\pi a_0^2$ | $8.7973554\times10^{-21}$ м² | $8.7973554\times10^{-17}$ см² |
| Сечение Томсона | $\sigma_\mathrm{T}=8\pi r_e^2/3$ | $6.6524587051\times10^{-29}$ м² | $6.6524587051\times10^{-25}$ см² |

$$
a_0=\frac{4\pi\varepsilon_0\hbar^2}{m_e e^2},\qquad
r_e=\frac{e^2}{4\pi\varepsilon_0m_ec^2},\qquad
\lambda_\mathrm{C}=2\pi\bar\lambda_\mathrm{C}.
$$

**$\pi a_0^2$ — масштаб площади**, а не универсальное сечение столкновения атомов: реальное сечение зависит от процесса и энергии.


### Атомные энергии и постоянная Ридберга

| Величина | Обозначение | Значение | Единица |
| --- | --- | --- | --- |
| Постоянная Ридберга | $R_\infty$ | $10\,973\,731.568157$ | м⁻¹ |
| Та же постоянная в СГС | $R_\infty$ | $109\,737.31568157$ | см⁻¹ |
| Энергия Ридберга | $E_\mathrm{Ry}=hcR_\infty$ | $13.605693122990$ | эВ |
| Энергия Хартри | $E_\mathrm{h}=2hcR_\infty$ | $27.211386245981$ | эВ |
| Энергия Ридберга в СИ | $E_\mathrm{Ry}$ | $2.1798723611030\times10^{-18}$ | Дж |
| Энергия Хартри в СИ | $E_\mathrm{h}$ | $4.3597447222060\times10^{-18}$ | Дж |

$$
R_\infty=\frac{m_e e^4}{8\varepsilon_0^2 c h^3}
=\frac{\alpha^2m_ec}{2h},\qquad
E_\mathrm{Ry}=\frac{\alpha^2m_ec^2}{2}.
$$

Индекс $\infty$ означает бесконечно тяжёлое ядро. Для спектра водородоподобного атома в нерелятивистском приближении учитывают приведённую массу: $R_M=R_\infty\,m_\mathrm{red}/m_e$, где $m_\mathrm{red}=m_eM/(m_e+M)$.

### Тепловое излучение

| Величина | Обозначение | СИ | СГС |
| --- | --- | --- | --- |
| Постоянная Стефана–Больцмана † | $\sigma$ | $5.670374419\ldots\times10^{-8}$ Вт/(м²·К⁴) | $5.670374419\ldots\times10^{-5}$ эрг/(см²·с·К⁴) |
| Первая радиационная постоянная † | $c_1=2\pi hc^2$ | $3.741771852\ldots\times10^{-16}$ Вт·м² | $3.741771852\ldots\times10^{-5}$ эрг·см²/с |
| Вторая радиационная постоянная † | $c_2=hc/k_\mathrm{B}$ | $1.438776877\ldots\times10^{-2}$ м·К | $1.438776877\ldots$ см·К |

Здесь $c_1$ используется для спектральной энергетической светимости; для спектральной яркости вместо $2\pi hc^2$ стоит $2hc^2$ на стерадиан. Закон $\lambda_\max T=b$ относится к максимуму спектра **по длине волны**.

### Нормальные условия и идеальный газ

| Величина | Обозначение | СИ | СГС |
| --- | --- | --- | --- |
| Температура $0\,{}^\circ\mathrm C$ † | $T_0$ | $273.15$ К | То же значение |
| Стандартная атмосфера † | $p_0$ | $101\,325$ Па | $1.01325\times10^6$ дин/см² |
| Концентрация идеального газа, число Лошмидта † | $n_0=p_0/(k_\mathrm{B}T_0)$ | $2.686780111\ldots\times10^{25}$ м⁻³ | $2.686780111\ldots\times10^{19}$ см⁻³ |
| Молярный объём идеального газа † | $V_m=RT_0/p_0$ | $22.41396954\ldots$ л/моль | $22\,413.96954\ldots$ см³/моль |

Здесь «нормальные условия» означают **273.15 К и 101325 Па**. При давлении **100000 Па** и той же температуре $V_m=22.71095464\ldots$ л/моль, а $n_0=2.651645804\ldots\times10^{25}$ м⁻³. Всегда проверяйте, какое давление принято в задаче.

Молярная масса сухого воздуха — примерно **28.97 г/моль**; это характеристика смеси, зависящая от её состава, а не фундаментальная постоянная.
