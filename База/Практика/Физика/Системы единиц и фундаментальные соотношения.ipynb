# Системы единиц и фундаментальные соотношения

Справочник переводов: сначала выберите область в оглавлении, затем найдите величину по названию или обозначению в поиске сверху. В каждой строке явно указано, **из какой единицы в какую** выполняется перевод.

**Нужна постоянная, а не перевод?** Значения $c$, $h$, $e$, масс частиц и атомных масштабов собраны в [физических константах](Физические%20константы.ipynb).

> **СГС — семейство систем.** Для механики достаточно сантиметра, грамма и секунды. Для электричества ниже применены электростатические единицы (статКл, статВ); для полей — гауссова СГС. Электрические и магнитные коэффициенты приведены с учебной точностью: не переносите формулы из СИ в СГС простой заменой названий единиц.

## 3. Электричество: заряд, потенциал и цепи

**статКл = statC**, **статВ = statV**, **статА = statA**. Числа с десятичной дробью ниже округлены. Для быстрой оценки можно использовать $1$ Кл $\approx3\times10^9$ статКл и $1$ статВ $\approx300$ В.

| Величина | Обозначение | В СИ | Эквивалент в электростатической СГС |
| --- | --- | --- | --- |
| Электрический заряд | $q$ | 1 Кл | $\approx2.99792458\times10^9$ статКл |
| Объёмная плотность заряда | $\rho_q$ | 1 Кл/м³ | $\approx2.99792458\times10^3$ статКл/см³ |
| Сила тока | $I$ | 1 А | $\approx2.99792458\times10^9$ статА |
| Плотность тока | $j$ | 1 А/м² | $\approx2.99792458\times10^5$ статА/см² |
| Потенциал, напряжение, ЭДС | $\varphi,U,\mathcal E$ | 1 В | $\approx3.335640952\times10^{-3}$ статВ |
| Электрическое поле | $E$ | 1 В/м | $\approx3.335640952\times10^{-5}$ статВ/см |
| Ёмкость | $C$ | 1 Ф | $\approx8.987551787\times10^{11}$ статФ (см) |
| Сопротивление, импеданс | $R,Z$ | 1 Ом | $\approx1.112650056\times10^{-12}$ статОм (с/см) |
| Проводимость | $G_\mathrm{el}=1/R$ | 1 См | $\approx8.987551787\times10^{11}$ см/с |
| Удельное сопротивление | $\rho_\mathrm{el}$ | 1 Ом·м | $\approx1.112650056\times10^{-10}$ с |
| Удельная проводимость | $\sigma_\mathrm{el}$ | 1 См/м | $\approx8.987551787\times10^9$ с⁻¹ |
| Индуктивность в электростатических единицах | $L_\mathrm{ESU}$ | 1 Гн | $\approx1.112650056\times10^{-12}$ статГн (с²/см) |

**См — сименс, см — сантиметр.** СтатГн относится к электростатической системе единиц; в электромагнитной СГС индуктивность выражают иначе. Всегда проверяйте систему в исходной формуле.

### Поляризация и электрическое смещение

В СИ: $\mathbf D=\varepsilon_0\mathbf E+\mathbf P$. В гауссовой СГС: $\mathbf D=\mathbf E+4\pi\mathbf P$.

| Величина | В СИ | В гауссовой СГС |
| --- | --- | --- |
| Поляризация $P$ | 1 Кл/м² | $\approx2.99792458\times10^5$ статКл/см² |
| Электрическое смещение $D$ | 1 Кл/м² | $\approx4\pi\times2.99792458\times10^5$ статКл/см² |
| Относительная диэлектрическая проницаемость | $\varepsilon_r=\varepsilon/\varepsilon_0$ | То же безразмерное значение для линейной изотропной среды |

Несмотря на одинаковые единицы $D$ и $P$, **их коэффициенты перевода отличаются на $4\pi$**. Абсолютную $\varepsilon$ в Ф/м нельзя отождествлять с безразмерной $\varepsilon_r$.


## 4. Магнетизм: B, H, момент и намагниченность

**Гс = G** (гаусс), **Э = Oe** (эрстед), **Мкс = Mx** (максвелл). Столбцы ниже связывают гауссову СГС и СИ; это стандартные учебные коэффициенты.

| Величина | Обозначение | В СИ | В СГС |
| --- | --- | --- | --- |
| Магнитная индукция | $B$ | 1 Тл | $10^4$ Гс |
| Напряжённость магнитного поля | $H$ | 1 А/м | $4\pi\times10^{-3}$ Э |
| Магнитный поток | $\Phi$ | 1 Вб | $10^8$ Мкс |
| Векторный потенциал | $A$ | 1 Вб/м | $10^6$ Гс·см |
| Магнитный момент | $\mathfrak m$ | 1 А·м² = 1 Дж/Тл | $10^3$ эрг/Гс = $10^3$ emu |
| Намагниченность | $M$ | 1 А/м | $10^{-3}$ emu/см³ |
| Магнитодвижущая сила | $\mathcal F$ | 1 А | $4\pi/10$ Гб (гильберт) |
| Магнитное сопротивление | $\mathcal R$ | 1 А/Вб | $4\pi\times10^{-9}$ Гб/Мкс |

### Не путать H и M

$$
\text{СИ:}\quad \mathbf B=\mu_0(\mathbf H+\mathbf M),
\qquad
\text{СГС:}\quad \mathbf B=\mathbf H+4\pi\mathbf M.
$$

В СИ $H$ и $M$ измеряются в А/м, но их коэффициенты перевода в СГС **различаются**. Запись «намагниченность в гауссах» может означать $4\pi M$, а не $M$ — проверяйте определение в источнике.

| Свойство линейной изотропной среды | СИ | Гауссова СГС |
| --- | --- | --- |
| Относительная магнитная проницаемость | $\mu_r=\mu/\mu_0$ | То же численное значение $\mu_r$ |
| Магнитная восприимчивость при $M=\chi H$ | $\chi_\mathrm{SI}$ | $\chi_\mathrm{G}=\chi_\mathrm{SI}/(4\pi)$ |


## 5. Перенос: вязкость, диффузия и тепло

| Величина | Обозначение | В СИ | В СГС |
| --- | --- | --- | --- |
| Динамическая вязкость | $\eta$ | 1 Па·с | 10 П (пуаз) |
| Кинематическая вязкость | $\nu_\mathrm{kin}=\eta/\rho$ | 1 м²/с | $10^4$ Ст (стокс) |
| Коэффициент диффузии | $D_\mathrm{diff}$ | 1 м²/с | $10^4$ см²/с |
| Теплопроводность | $\kappa$ | 1 Вт/(м·К) | $10^5$ эрг/(см·с·К) |

Не путайте динамическую вязкость $\eta$ с кинематической $\nu_\mathrm{kin}$. В теплопроводности температура входит в размерность: $[\kappa]=MLT^{-3}\Theta^{-1}$.

## 7. Масса и энергия покоя

Масса выражается в **эВ/$c^2$**, энергия — в **эВ**. Их связывает $E_0=mc^2$.

| Соотношение | Значение |
| --- | --- |
| 1 эВ/$c^2$ | $\approx1.782661922\times10^{-36}$ кг |
| 1 МэВ/$c^2$ | $\approx1.782661922\times10^{-30}$ кг |
| 1 ГэВ/$c^2$ | $\approx1.782661922\times10^{-27}$ кг |
| 1 кг | $\approx5.609588604\times10^{35}$ эВ/$c^2$ |
| 1 атомная единица массы (u, Da) | $\approx931.49410372$ МэВ/$c^2$ |

Например, масса электрона $m_e\approx0.51099895$ МэВ/$c^2$, а его энергия покоя $m_ec^2\approx0.51099895$ МэВ.


## 8. Фотоны: энергия, частота и длина волны

$$
E=h\nu=\hbar\omega=\frac{hc}{\lambda}=hc\tilde\nu,
\qquad \omega=2\pi\nu,\quad \tilde\nu=\frac1\lambda,\quad k=\frac{2\pi}{\lambda}.
$$

Здесь $\lambda$ — длина волны **в вакууме**. Спектроскопическое волновое число $\tilde\nu$ и угловое волновое число $k$ отличаются в $2\pi$ раз.

| Связь | Практическая запись |
| --- | --- |
| Энергия → частота | $\nu\,[\mathrm{Hz}]\approx2.417989242\times10^{14}\,E\,[\mathrm{eV}]$ |
| Энергия → угловая частота | $\omega\,[\mathrm{s}^{-1}]\approx1.519267448\times10^{15}\,E\,[\mathrm{eV}]$ |
| Энергия ↔ длина волны | $E\,[\mathrm{eV}]\,\lambda\,[\mathrm{nm}]\approx1239.841984$ |
| Энергия → волновое число | $\tilde\nu\,[\mathrm{cm}^{-1}]\approx8065.543937\,E\,[\mathrm{eV}]$ |
| 1 см⁻¹ → энергия | $\approx1.239841984\times10^{-4}$ эВ |
| 1 м⁻¹ → энергия | $\approx1.239841984\times10^{-6}$ эВ |

**Пример:** фотон с длиной волны 500 нм имеет энергию $1239.841984/500\approx2.47968$ эВ.


## 9. Температура и тепловая энергия

$$
E_T=k_\mathrm{B}T,\qquad k_\mathrm{B}\approx8.617333262\times10^{-5}\ \mathrm{eV/K}.
$$

| Соотношение | Значение |
| --- | --- |
| 1 К → $k_\mathrm{B}T$ | $\approx8.617333262\times10^{-5}$ эВ |
| 1 эВ → температура при $E_T=k_\mathrm{B}T$ | $\approx11604.51812$ К |
| При 300 К | $k_\mathrm{B}T\approx0.0258520$ эВ = 25.8520 мэВ |
| Цельсий → кельвин | $T\,[\mathrm K]=t\,[{}^\circ\mathrm C]+273.15$ |

$k_\mathrm{B}T$ — энергетический масштаб. Например, средняя поступательная энергия частицы классического одноатомного идеального газа равна $3k_\mathrm{B}T/2$, а не $k_\mathrm{B}T$.


## 10. Атомные и естественные единицы

| Единица или комбинация | Эквивалент |
| --- | --- |
| 1 боровский радиус $a_0$ | $\approx0.05291772105$ нм |
| 1 хартри $E_\mathrm h$ | $\approx27.211386246$ эВ |
| 1 ридберг $E_\mathrm{Ry}=E_\mathrm h/2$ | $\approx13.605693123$ эВ |
| $\hbar c$ | $\approx197.3269804$ МэВ·фм = $197.3269804$ эВ·нм |
| $hc$ | $\approx1239.841984$ эВ·нм |
| $\hbar$ | $\approx6.582119569\times10^{-16}$ эВ·с |
| 1 ГэВ⁻¹ в единицах длины | $\approx0.1973269804$ фм, после умножения на $\hbar c$ |
| 1 ГэВ⁻¹ в единицах времени | $\approx6.582119569\times10^{-25}$ с, после умножения на $\hbar$ |
| 1 ГэВ⁻² в единицах сечения | $\approx0.3893793721$ мб, после умножения на $(\hbar c)^2$ |

В естественных единицах $\hbar=c=1$ энергия, масса и импульс имеют одну размерность; длина и время — обратную энергию. В атомных единицах обычно полагают $m_e=e=\hbar=4\pi\varepsilon_0=1$. Это **разные соглашения**.


## 11. Размерности: быстрая проверка формулы

$M$ — масса, $L$ — длина, $T$ — время, $Q$ — заряд, $\Theta$ — температура. Это символы размерностей, а не обозначения конкретных единиц. В СИ заряд можно выразить через ток: $Q=IT$.

### Механика и перенос

| Величина | Размерность |
| --- | --- |
| Скорость; ускорение | $LT^{-1}$; $LT^{-2}$ |
| Частота; вихревость | $T^{-1}$ |
| Массовая плотность | $ML^{-3}$ |
| Импульс; плотность импульса | $MLT^{-1}$; $ML^{-2}T^{-1}$ |
| Сила | $MLT^{-2}$ |
| Работа, энергия | $ML^2T^{-2}$ |
| Мощность; её объёмная плотность | $ML^2T^{-3}$; $ML^{-1}T^{-3}$ |
| Давление; плотность энергии | $ML^{-1}T^{-2}$ |
| Динамическая вязкость | $ML^{-1}T^{-1}$ |
| Диффузия; кинематическая вязкость | $L^2T^{-1}$ |
| Теплопроводность | $MLT^{-3}\Theta^{-1}$ |

### Электрические и магнитные величины в СИ

| Величина | Размерность |
| --- | --- |
| Заряд; плотность заряда | $Q$; $QL^{-3}$ |
| Ток; плотность тока | $QT^{-1}$; $QL^{-2}T^{-1}$ |
| Напряжение, потенциал, ЭДС | $ML^2T^{-2}Q^{-1}$ |
| Электрическое поле | $MLT^{-2}Q^{-1}$ |
| Поляризация; электрическое смещение | $QL^{-2}$ |
| Ёмкость | $Q^2T^2M^{-1}L^{-2}$ |
| Сопротивление, импеданс | $ML^2T^{-1}Q^{-2}$ |
| Проводимость | $Q^2TM^{-1}L^{-2}$ |
| Удельное сопротивление | $ML^3T^{-1}Q^{-2}$ |
| Удельная проводимость | $Q^2TM^{-1}L^{-3}$ |
| Индуктивность | $ML^2Q^{-2}$ |
| Магнитная индукция | $MT^{-1}Q^{-1}$ |
| Магнитный поток; векторный потенциал | $ML^2T^{-1}Q^{-1}$; $MLT^{-1}Q^{-1}$ |
| Напряжённость магнитного поля; намагниченность | $QL^{-1}T^{-1}$ |
| Магнитный момент | $QL^2T^{-1}$ |

| Абсолютная магнитная проницаемость | $MLQ^{-2}$ |

### Что меняется в гауссовой СГС

Из закона Кулона $F=q_1q_2/r^2$ следует $[q]=M^{1/2}L^{3/2}T^{-1}$. Поэтому электрические величины выражаются через $M,L,T$; в магнитных формулах необходимо также учитывать явные множители $c$.

| Величина | Размерность в гауссовой СГС |
| --- | --- |
| Заряд | $M^{1/2}L^{3/2}T^{-1}$ |
| Ток | $M^{1/2}L^{3/2}T^{-2}$ |
| Потенциал | $M^{1/2}L^{1/2}T^{-1}$ |
| Поля $E,D,B,H$; поляризация и намагниченность | $M^{1/2}L^{-1/2}T^{-1}$ |
| Ёмкость; сопротивление | $L$; $TL^{-1}$ |
| Удельная проводимость; удельное сопротивление | $T^{-1}$; $T$ |
| Магнитный поток | $M^{1/2}L^{3/2}T^{-1}$ |
| Векторный потенциал | $M^{1/2}L^{1/2}T^{-1}$ |
| Магнитный момент | $M^{1/2}L^{5/2}T^{-1}$ |

Равенство размерностей — необходимая проверка формулы, но оно не определяет безразмерные коэффициенты: так нельзя найти пропущенные $2$, $\pi$ или $4\pi$.
