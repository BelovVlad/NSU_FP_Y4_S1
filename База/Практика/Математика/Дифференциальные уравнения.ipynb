# Дифференциальные уравнения — справочник

Компактный справочник по основным методам решения ОДУ, линейным системам, вариационному исчислению, малым колебаниям, периодическим решениям, устойчивости, фазовым портретам и УЧП первого порядка.

> **Как пользоваться:** в основном тексте оставлены формулы, критерии и алгоритмы.  
> Примеры и идеи доказательств скрыты и раскрываются нажатием.  
> Основа — существующий черновик справочника и конспект лекций по ДФУ. Очевидные опечатки исходного черновика исправлены.

Обозначения:
- $E$ — единичная матрица;
- $\Phi(t)$ — фундаментальная матрица решений;
- $A^*$ — транспонированная матрица (для вещественных матриц $A^*=A^T$);
- $\operatorname{Re}\lambda$ — вещественная часть собственного значения;
- $\dot y=dy/dt$, $y'=dy/dx$.


# 0. Быстрый выбор метода

| Вид задачи | Что делать |
|---|---|
| $y'=f(x)g(y)$ | Разделить переменные |
| $y'=\varphi(y/x)$ | $y=xz$, затем разделить переменные |
| $y'=f(\alpha x+\beta y+\gamma)$ | $z=\alpha x+\beta y+\gamma$ |
| $y'+p(x)y=q(x)$ | Линейное уравнение, интегрирующий множитель |
| $y'+p(x)y=q(x)y^\alpha$ | Бернулли: $z=y^{1-\alpha}$ |
| $y'=a(x)y^2+b(x)y+c(x)$ и известно $y_p$ | Риккати: $y=y_p+1/u$ |
| $P\,dx+Q\,dy=0$, $P_y=Q_x$ | Полный дифференциал $dU=0$ |
| $Y'=A(t)Y+F(t)$ | Фундаментальная матрица + вариация постоянных |
| $M\ddot x+Kx=f(t)$ | Нормальные моды: $\det(K-\lambda M)=0$ |
| $I[y]=\int F(x,y,y')dx\to\mathrm{ext}$ | Уравнение Эйлера–Лагранжа |
| $\dot y=f(y)$, исследование равновесия | Якобиан $Df(c)$ / функция Ляпунова |
| $u_t+\sum f_i u_{y_i}=0$ | Характеристики / первые интегралы |

> Если несколько методов подходят одновременно, обычно выбирают тот, который сильнее упрощает уравнение.


# 1. ОДУ первого порядка

Общий вид:
$$
y'=f(x,y).
$$

Задача Коши:
$$
y'=f(x,y),\qquad y(x_0)=y_0.
$$

При условиях теоремы Пикара (достаточно непрерывности $f$ и локальной непрерывности $f_y$) решение задачи Коши локально существует и единственно.


## 1.1. Уравнения с разделяющимися переменными

Если
$$
y'=f(x)g(y),
$$
то при $g(y)\neq0$
$$
\frac{dy}{g(y)}=f(x)\,dx,
$$
откуда
$$
\boxed{\int\frac{dy}{g(y)}=\int f(x)\,dx+C.}
$$

Отдельно проверить стационарные решения из $g(y)=0$.

<details>
<summary><b>Пример решения: $y'=xy$</b></summary>
<div>

<p>Разделяем переменные:</p>
<div>$$\frac{dy}{y}=x\,dx.$$</div>
<p>Интегрируем:</p>
<div>$$\ln|y|=\frac{x^2}{2}+C.$$</div>
<div>$$y=C_1e^{x^2/2}.$$</div>

</div>
</details>

## 1.2. Однородные уравнения первого порядка

Если
$$
y'=f(x,y),\qquad f(\lambda x,\lambda y)=f(x,y),
$$
то обычно
$$
y'=\varphi\!\left(\frac yx\right).
$$

Подстановка:
$$
y=xz,\qquad y'=z+xz'.
$$

Получаем:
$$
\boxed{\frac{dz}{\varphi(z)-z}=\frac{dx}{x}.}
$$

<details>
<summary><b>Пример решения: $y'=1+y/x$</b></summary>
<div>

<p>Положим $y=xz$. Тогда $z+xz'=1+z$, откуда $xz'=1$.</p>
<div>$$z=\ln|x|+C,\qquad \boxed{y=x(\ln|x|+C).}$$</div>

</div>
</details>

## 1.3. Уравнения вида $y'=f(\alpha x+\beta y+\gamma)$

Подстановка:
$$
z=\alpha x+\beta y+\gamma.
$$

Тогда
$$
z'=\alpha+\beta f(z).
$$

Получаем:
$$
\boxed{\frac{dz}{\alpha+\beta f(z)}=dx.}
$$

Отдельно проверить постоянные значения $z$, для которых $\alpha+\beta f(z)=0$.


## 1.4. Линейное уравнение первого порядка

Стандартная форма:
$$
\boxed{y'+p(x)y=q(x).}
$$

Интегрирующий множитель:
$$
\mu(x)=e^{\int p(x)\,dx}.
$$

После умножения:
$$
(\mu y)'=\mu q.
$$

Итог:
$$
\boxed{
y(x)=\frac{1}{\mu(x)}
\left(C+\int \mu(x)q(x)\,dx\right).
}
$$

<details>
<summary><b>Пример решения: $y'-2y=e^x$</b></summary>
<div>

<p>$\mu=e^{-2x}$, поэтому $(e^{-2x}y)'=e^{-x}$.</p>
<div>$$e^{-2x}y=-e^{-x}+C,$$</div>
<div>$$\boxed{y=-e^x+Ce^{2x}.}$$</div>

</div>
</details>

## 1.5. Уравнение Бернулли

$$
y'+p(x)y=q(x)y^\alpha,\qquad \alpha\neq0,1.
$$

Подстановка:
$$
z=y^{1-\alpha}.
$$

Получаем:
$$
\boxed{
z'+(1-\alpha)p(x)z=(1-\alpha)q(x).
}
$$

Если $\alpha>0$, отдельно проверить $y=0$.


## 1.6. Уравнение Риккати

$$
y'=a(x)y^2+b(x)y+c(x).
$$

Если известно частное решение $y_p(x)$:
$$
y=y_p+\frac1u.
$$

Тогда:
$$
\boxed{
u'+(2ay_p+b)u=-a.
}
$$

Эквивалентно $s=y-y_p$ даёт:
$$
s'=(2ay_p+b)s+as^2.
$$

<details>
<summary><b>Пример: $y'=(y-1)^2$</b></summary>
<div>

<p>$y_p=1$. Положим $y=1+1/u$.</p>
<div>$$-\frac{u'}{u^2}=\frac1{u^2}\Rightarrow u'=-1.$$</div>
<div>$$\boxed{y=1+\frac1{C-x}.}$$</div>
<p>Отдельное решение: $y=1$.</p>

</div>
</details>

## 1.7. Симметричная форма и полный дифференциал

$$
P(x,y)\,dx+Q(x,y)\,dy=0.
$$

Если $Q\neq0$:
$$
\frac{dy}{dx}=-\frac{P}{Q}.
$$

Если $P\neq0$:
$$
\frac{dx}{dy}=-\frac{Q}{P}.
$$

Если существует $U(x,y)$:
$$
U_x=P,\qquad U_y=Q,
$$
то
$$
\boxed{U(x,y)=C.}
$$

На односвязной области:
$$
\boxed{P_y=Q_x.}
$$

Алгоритм:
1. $U=\int P\,dx+\varphi(y)$;
2. найти $U_y$;
3. сравнить с $Q$.

<details>
<summary><b>Пример решения</b></summary>
<div>

<div>$$(2xy+1)\,dx+(x^2+2y)\,dy=0.$$</div>
<p>$P_y=Q_x=2x$.</p>
<div>$$U=x^2y+x+\varphi(y),\qquad \varphi=y^2.$$</div>
<div>$$\boxed{x^2y+x+y^2=C.}$$</div>

</div>
</details>

## 1.8. Интегрирующий множитель

Если
$$
P\,dx+Q\,dy=0,\qquad P_y\neq Q_x,
$$
ищем $\mu$, чтобы $(\mu P)_y=(\mu Q)_x$.

Если
$$
\frac{P_y-Q_x}{Q}
$$
зависит только от $x$:
$$
\boxed{
\mu(x)=\exp\!\left(\int\frac{P_y-Q_x}{Q}\,dx\right).
}
$$

Если
$$
\frac{Q_x-P_y}{P}
$$
зависит только от $y$:
$$
\boxed{
\mu(y)=\exp\!\left(\int\frac{Q_x-P_y}{P}\,dy\right).
}
$$


# 2. Линейные ОДУ и системы

Этот раздел — рабочий мост к колебаниям, периодическим решениям и устойчивости.


## 2.1. Линейное ОДУ порядка $n$

$$
a_n(x)y^{(n)}+\cdots+a_1(x)y'+a_0(x)y=f(x).
$$

Для постоянных коэффициентов однородного уравнения:
$$
\boxed{
a_n\lambda^n+\cdots+a_1\lambda+a_0=0.
}
$$

Для корня $\lambda$ кратности $m$:
$$
e^{\lambda x},\ xe^{\lambda x},\ldots,x^{m-1}e^{\lambda x}.
$$

Для пары $\lambda=\alpha\pm i\beta$:
$$
e^{\alpha x}\cos\beta x,\qquad
e^{\alpha x}\sin\beta x.
$$

Для неоднородного уравнения:
$$
\boxed{y=y_{\mathrm{hom}}+y_{\mathrm{part}}.}
$$


## 2.2. Линейная система и фундаментальная матрица

$$
\boxed{Y'=A(x)Y+F(x).}
$$

$$
\Phi'=A\Phi,\qquad \det\Phi\neq0.
$$

$$
\boxed{Y_h=\Phi(x)C.}
$$

Вариация постоянных:
$$
\boxed{
Y(x)=\Phi(x)\left[
C+\int_{x_0}^{x}\Phi^{-1}(s)F(s)\,ds
\right].
}
$$

Если $A=\mathrm{const}$:
$$
\boxed{\Phi(x)=e^{xA}.}
$$

Для $Y(x_0)=Y_0$:
$$
\boxed{
Y(x)=e^{(x-x_0)A}Y_0+
\int_{x_0}^{x}e^{(x-s)A}F(s)\,ds.
}
$$


## 2.3. Сведение уравнения высокого порядка к системе

Для
$$
y^{(n)}+a_{n-1}(t)y^{(n-1)}+\cdots+a_0(t)y=f(t)
$$
вводим
$$
y_1=y,\quad y_2=y',\quad\ldots,\quad y_n=y^{(n-1)}.
$$

Тогда
$$
\dot{\mathbf y}=A_n(t)\mathbf y+\mathbf f(t),
$$
где
$$
A_n(t)=
\begin{pmatrix}
0&1&0&\cdots&0\\
0&0&1&\cdots&0\\
\vdots&&&\ddots&\vdots\\
0&0&0&\cdots&1\\
-a_0&-a_1&-a_2&\cdots&-a_{n-1}
\end{pmatrix}.
$$


# 3. Вариационное исчисление

Типичная задача:
$$
I[y]=\int_{x_0}^{x_1}F(x,y,y')\,dx\to\operatorname{ext},
$$
при краевых условиях
$$
y(x_0)=y_0,\qquad y(x_1)=y_1.
$$


## 3.1. Уравнение Эйлера–Лагранжа

Необходимое условие локального экстремума:
$$
\boxed{
\frac{\partial F}{\partial y}
-\frac{d}{dx}\frac{\partial F}{\partial y'}=0.
}
$$

Функция, удовлетворяющая этому уравнению и краевым условиям, называется экстремалью.

<details>
<summary><b>Идея вывода</b></summary>
<div>

<p>Берётся вариация $y=\tilde y+\varepsilon\eta$, где $\eta(x_0)=\eta(x_1)=0$.</p>
<p>Для $g(\varepsilon)=I[\tilde y+\varepsilon\eta]$ в точке экстремума $g'(0)=0$.</p>
<p>После дифференцирования и интегрирования по частям:</p>
<div>$$
\int_{x_0}^{x_1}\left(F_y-\frac{d}{dx}F_{y'}\right)\eta\,dx=0.
$$</div>
<p>Основная лемма вариационного исчисления даёт уравнение Эйлера.</p>

</div>
</details>

<details>
<summary><b>Пример: квадратичный функционал</b></summary>
<div>

<div>$$I[y]=\int_0^1\bigl((y')^2+y^2\bigr)\,dx.$$</div>
<p>$F_y=2y$, $F_{y'}=2y'$.</p>
<div>$$2y-2y''=0\quad\Rightarrow\quad\boxed{y''-y=0}.$$</div>

</div>
</details>

## 3.2. Первые интегралы уравнения Эйлера

Если $F$ не зависит от $y$:
$$
\boxed{
\frac{\partial F}{\partial y'}=C.
}
$$

Если $F$ не зависит явно от $x$:
$$
F=F(y,y'),
$$
то
$$
\boxed{
F-y'\frac{\partial F}{\partial y'}=C.
}
$$

Вторая формула особенно полезна для брахистохроны и поверхности вращения минимальной площади.


## 3.3. Обобщения уравнения Эйлера

### Несколько функций
$$
I[y_1,\ldots,y_n]=
\int F(x,y_1,\ldots,y_n,y_1',\ldots,y_n')\,dx,
$$
$$
\boxed{
F_{y_j}-\frac{d}{dx}F_{y_j'}=0,\qquad j=1,\ldots,n.
}
$$

### Высшие производные
$$
I[y]=\int F(x,y,y',\ldots,y^{(n)})\,dx,
$$
$$
\boxed{
\sum_{k=0}^{n}(-1)^k\frac{d^k}{dx^k}
\frac{\partial F}{\partial y^{(k)}}=0.
}
$$

### Несколько независимых переменных
$$
I[z]=\iint_DF(x,y,z,z_x,z_y)\,dx\,dy,
$$
$$
\boxed{
F_z-\frac{\partial}{\partial x}F_{z_x}
-\frac{\partial}{\partial y}F_{z_y}=0.
}
$$


## 3.4. Принцип стационарного действия

$$
L=T-U,\qquad
S=\int_{t_0}^{t_1}L(q,\dot q,t)\,dt.
$$

Уравнения движения:
$$
\boxed{
\frac{d}{dt}\frac{\partial L}{\partial\dot q_i}
-\frac{\partial L}{\partial q_i}=0.
}
$$

Если $L$ не зависит явно от времени:
$$
\boxed{
L-\sum_i\dot q_i\frac{\partial L}{\partial\dot q_i}=C.
}
$$


## 3.5. Условный экстремум и изопериметрические задачи

Если
$$
I[y]=\int F\,dx
$$
экстремизируется при
$$
J[y]=\int G\,dx=\mathrm{const},
$$
строим
$$
\boxed{\widetilde F=F+\lambda G.}
$$

Для ограничения
$$
G(t,y_1,\ldots,y_n)=0
$$
используем
$$
\boxed{\widetilde F=F+\lambda(t)G.}
$$

<details>
<summary><b>Классическая изопериметрическая задача</b></summary>
<div>

<p>Среди замкнутых кривых заданной длины ищется кривая максимальной площади.</p>
<p>После введения множителя Лагранжа:</p>
<div>$$(x-c_2)^2+(y-c_1)^2=\lambda^2.$$</div>
<p>Экстремаль — окружность.</p>

</div>
</details>

## 3.6. Классические результаты

<details>
<summary><b>Брахистохрона</b></summary>
<div>

<div>$$
T[y]=\int
\frac{\sqrt{1+(y')^2}}
{\sqrt{2g(y_0-y)}}\,dx.
$$</div>
<p>Так как $F$ не зависит явно от $x$, применяется первый интеграл.</p>
<p>Результат — циклоида:</p>
<div>$$
x(t)=x_0+\frac{c}{2}(t-\sin t),\qquad
y(t)=y_0-\frac{c}{2}(1-\cos t).
$$</div>

</div>
</details>

<details>
<summary><b>Поверхность вращения наименьшей площади</b></summary>
<div>

<div>$$
S[y]=\int 2\pi y\sqrt{1+(y')^2}\,dx.
$$</div>
<div>$$
\frac{2\pi y}{\sqrt{1+(y')^2}}=C.
$$</div>
<div>$$
\boxed{
y(x)=c_1\cosh\!\left(\frac{x+c_2}{c_1}\right).
}
$$</div>
<p>Получается цепная линия; поверхность вращения — катеноид.</p>

</div>
</details>

<details>
<summary><b>Геодезическая на сфере</b></summary>
<div>

<p>Ограничение:</p>
<div>$$x^2+y^2+z^2=R^2.$$</div>
<p>Экстремаль лежит в плоскости, проходящей через центр сферы, поэтому геодезическая — дуга большой окружности.</p>

</div>
</details>

# 4. Системы малых колебаний

$$
\boxed{
M\ddot{\mathbf x}+K\mathbf x=0,
}
$$
где
$$
M=M^T>0,\qquad K=K^T\ge0.
$$

$M$ — матрица масс, $K$ — матрица жёсткостей.


## 4.1. Нормальные моды

Обобщённая спектральная задача:
$$
\boxed{
\det(K-\lambda M)=0,
}
$$
$$
\boxed{
(K-\lambda_jM)\mathbf v_j=0.
}
$$

При $M>0$, $K\ge0$:
$$
\lambda_j\ge0.
$$

Разложение:
$$
\boxed{
\mathbf x(t)=\sum_{j=1}^n q_j(t)\mathbf v_j.
}
$$

Для каждой моды:
$$
\boxed{
q_j''+\lambda_jq_j=0.
}
$$

Собственные частоты:
$$
\boxed{
\omega_j=\sqrt{\lambda_j}.
}
$$

Если $\lambda_j>0$:
$$
q_j=A_j\cos(\omega_jt)+B_j\sin(\omega_jt).
$$

Если $\lambda_j=0$:
$$
q_j=A_jt+B_j.
$$

Для различных $\lambda_i\neq\lambda_j$:
$$
(M\mathbf v_i,\mathbf v_j)=0.
$$

<details>
<summary><b>Пример: две связанные моды</b></summary>
<div>

<p>Пусть $M=E$ и</p>
<div>$$
K=\begin{pmatrix}2&-1\\-1&2\end{pmatrix}.
$$</div>
<p>$\lambda_1=1$, $\lambda_2=3$.</p>
<div>$$
v_1=\binom11,\qquad v_2=\binom1{-1},
$$</div>
<div>$$\omega_1=1,\qquad \omega_2=\sqrt3.$$</div>

</div>
</details>

## 4.2. Вынужденные малые колебания

Для
$$
M\ddot{\mathbf x}+K\mathbf x=\mathbf f(t)
$$
выбираем $M$-ортогональный базис собственных векторов $\mathbf w_j$.

$$
\mathbf x(t)=\sum_jq_j(t)\mathbf w_j.
$$

Тогда:
$$
\boxed{
q_j''+\lambda_jq_j=\widetilde f_j(t),
}
$$
где
$$
\boxed{
\widetilde f_j(t)=
\frac{(\mathbf f(t),\mathbf w_j)}
{(M\mathbf w_j,\mathbf w_j)}.
}
$$

При кратном собственном значении внутри собственного подпространства можно выполнить $M$-ортогонализацию Грама–Шмидта.


# 5. Зависимость решения от параметров и начальных данных

Рассматриваем
$$
y'=f(t,y,\mu),\qquad y(t_0)=y_0.
$$

Главная идея: на любом фиксированном компактном промежутке внутри интервала существования базового решения малое изменение параметра/начального условия даёт малое изменение решения.

> Это утверждение нельзя автоматически переносить на бесконечный промежуток времени.


## 5.1. Непрерывная зависимость

Если решение при $\mu=\mu_*$ существует на $(\alpha,\omega)$, то для любого
$$
[t_1,t_2]\subset(\alpha,\omega)
$$
существует $\Delta>0$, такое что при $|\mu-\mu_*|<\Delta$ соседние решения существуют на всём $[t_1,t_2]$ и
$$
\boxed{
y(t,\mu)\to y(t,\mu_*)
}
$$
на этом фиксированном промежутке.

Аналогично для зависимости от $y_0$.


## 5.2. Дифференцируемость по параметру

Пусть
$$
y'=f(t,y,\mu),\qquad y(t_0)=y_0,
$$
и $y_0$ не зависит от $\mu$.

Обозначим
$$
v(t,\mu)=\frac{\partial y}{\partial\mu}.
$$

Тогда:
$$
\boxed{
\frac{\partial v}{\partial t}
=
f_y(t,y,\mu)v+f_\mu(t,y,\mu),
\qquad
v(t_0,\mu)=0.
}
$$

Линейное приближение:
$$
\boxed{
y(t,\mu)
=
y(t,\mu_*)
+
\frac{\partial y}{\partial\mu}(t,\mu_*)
(\mu-\mu_*)
+
o(\mu-\mu_*).
}
$$

<details>
<summary><b>Пример из лекций</b></summary>
<div>

<p>Для</p>
<div>$$
y'=4t\mu-y^2,\qquad y(1)=1
$$</div>
<p>при $\mu=0$:</p>
<div>$$y(t,0)=\frac1t.$$</div>
<p>Пусть $w(t)=\left.\partial y/\partial\mu\right|_{\mu=0}$. Тогда</p>
<div>$$
w'=4t-\frac{2}{t}w,\qquad w(1)=0.
$$</div>
<div>$$
w(t)=t^2-\frac1{t^2}.
$$</div>
<div>$$
y(t,\mu)=\frac1t+
\left(t^2-\frac1{t^2}\right)\mu+o(\mu).
$$</div>

</div>
</details>

## 5.3. Дифференцируемость по начальному условию

Для
$$
y'=f(t,y),\qquad y(t_0)=y_0
$$
обозначим
$$
v(t,y_0)=\frac{\partial y}{\partial y_0}.
$$

Тогда:
$$
\boxed{
v_t=f_y(t,y)v,
\qquad
v(t_0)=1.
}
$$

Для систем это переходит в матричное вариационное уравнение с якобианом правой части.


# 6. Периодические решения

Рассмотрим
$$
\dot{\mathbf y}=A(t)\mathbf y+\mathbf f(t),
$$
где
$$
A(t+T)=A(t),\qquad
\mathbf f(t+T)=\mathbf f(t).
$$


## 6.1. Критерий периодичности

Решение $T$-периодично тогда и только тогда, когда
$$
\boxed{
\mathbf y(0)=\mathbf y(T).
}
$$

Пусть $\Phi(t)$ — фундаментальная матрица однородной системы.

Единственное $T$-периодическое решение существует тогда и только тогда, когда
$$
\boxed{
\det\bigl(\Phi(T)-\Phi(0)\bigr)\neq0.
}
$$

Если $\Phi(0)=E$:
$$
\boxed{
\det(\Phi(T)-E)\neq0.
}
$$

Если определитель равен нулю, условие периодичности сводится к
$$
B\mathbf c=\boldsymbol\psi.
$$

- $\operatorname{rank}B=\operatorname{rank}(B|\psi)$ — бесконечно много решений;
- ранги различны — решений нет.

<details>
<summary><b>Идея критерия</b></summary>
<div>

<p>Общее решение:</p>
<div>$$
\mathbf y(t)=
\Phi(t)\mathbf c+
\Phi(t)\int_0^t\Phi^{-1}(s)\mathbf f(s)\,ds.
$$</div>
<p>Условие $\mathbf y(0)=\mathbf y(T)$ превращается в линейное уравнение относительно $\mathbf c$.</p>

</div>
</details>

## 6.2. Постоянная матрица

Если $A(t)=A=\mathrm{const}$:
$$
\Phi(t)=e^{tA}.
$$

Если $\lambda_j$ — собственные значения $A$, то единственное $T$-периодическое решение существует тогда и только тогда, когда
$$
\boxed{
e^{\lambda_jT}\neq1
\qquad \forall j.
}
$$

Эквивалентно:
$$
\lambda_jT\neq2\pi ik,\qquad k\in\mathbb Z.
$$


## 6.3. ОДУ высокого порядка

Для
$$
y^{(n)}+a_{n-1}(t)y^{(n-1)}+\cdots+a_0(t)y=f(t),
$$
где коэффициенты и $f$ имеют период $T$, решение периодично тогда и только тогда, когда
$$
\boxed{
y^{(k)}(0)=y^{(k)}(T),
\qquad k=0,\ldots,n-1.
}
$$

Для постоянных коэффициентов критерий единственности:
$$
\boxed{
e^{\lambda_jT}\neq1.
}
$$


## 6.4. Метод Фурье для $y''+\alpha y'+\beta y=f(t)$

Пусть
$$
f(t)=\frac{a_0}{2}
+\sum_{k=1}^{\infty}
\left(a_k\cos\omega_kt+b_k\sin\omega_kt\right),
\qquad
\omega_k=\frac{2\pi k}{T}.
$$

Ищем
$$
y_k=A_k\cos\omega_kt+B_k\sin\omega_kt.
$$

Обозначим
$$
d_k=\beta-\omega_k^2.
$$

Тогда:
$$
\begin{pmatrix}
d_k&\alpha\omega_k\\
-\alpha\omega_k&d_k
\end{pmatrix}
\binom{A_k}{B_k}
=
\binom{a_k}{b_k}.
$$

Если
$$
D_k=d_k^2+\alpha^2\omega_k^2\neq0,
$$
то
$$
\boxed{
A_k=\frac{d_ka_k-\alpha\omega_kb_k}{D_k},
\qquad
B_k=\frac{\alpha\omega_ka_k+d_kb_k}{D_k}.
}
$$

При $\beta\neq0$ нулевая гармоника:
$$
y_0=\frac{a_0}{2\beta}.
$$

> Если $D_k=0$, нужно отдельно проверять совместность: это резонансная/вырожденная ситуация.


# 7. Теория устойчивости

Рассматриваем
$$
\dot{\mathbf y}=\mathbf f(t,\mathbf y)
$$
и некоторое решение $\mathbf y_*(t)$.


## 7.1. Основные определения

### Устойчивость по Ляпунову

$\mathbf y_*(t)$ устойчиво, если для любого $\varepsilon>0$ существует $\delta>0$ такое, что
$$
\|\mathbf y(t_0)-\mathbf y_*(t_0)\|<\delta
$$
влечёт
$$
\boxed{
\|\mathbf y(t)-\mathbf y_*(t)\|<\varepsilon
\qquad \forall t\ge t_0,
}
$$
и соседнее решение существует при всех $t\ge t_0$.

### Асимптотическая устойчивость

Кроме устойчивости:
$$
\boxed{
\|\mathbf y(t)-\mathbf y_*(t)\|\to0,
\qquad t\to+\infty.
}
$$

### Сведение к нулевому решению

Положим
$$
\mathbf z=\mathbf y-\mathbf y_*.
$$

Исследование устойчивости $\mathbf y_*$ сводится к устойчивости нулевого решения для уравнения в $\mathbf z$.


## 7.2. Линейная система с постоянной матрицей

Для
$$
\dot{\mathbf y}=A\mathbf y
$$
главную роль играет спектр $A$.

### Асимптотическая устойчивость
$$
\boxed{
\operatorname{Re}\lambda_j<0\quad\forall j
\iff
\mathbf y=0\ \text{асимптотически устойчиво}.
}
$$

### Неустойчивость
Если
$$
\boxed{
\exists\,\lambda_k:\operatorname{Re}\lambda_k>0,
}
$$
то нулевое решение неустойчиво.

### Граница $\operatorname{Re}\lambda=0$

Если $\operatorname{Re}\lambda_j\le0$ для всех $j$ и есть собственные значения с нулевой вещественной частью:

- если у них нет присоединённых векторов (жордановы блоки размера $1$), нулевое решение устойчиво по Ляпунову;
- если есть присоединённый вектор, нулевое решение неустойчиво.

> Чисто мнимый спектр сам по себе не даёт асимптотической устойчивости.


## 7.3. Функция Ляпунова

Для автономной системы
$$
\dot{\mathbf y}=\mathbf f(\mathbf y),\qquad \mathbf f(0)=0
$$
ищем функцию $V(\mathbf y)$.

Условия:
$$
V(0)=0,\qquad V(\mathbf y)>0\quad(\mathbf y\neq0).
$$

Производная вдоль траектории:
$$
\boxed{
\dot V=
\nabla V(\mathbf y)\cdot\mathbf f(\mathbf y).
}
$$

Критерии:
- $\dot V\le0$ — устойчивость по Ляпунову;
- $\dot V<0$ при $\mathbf y\neq0$ — асимптотическая устойчивость;
- в формулировке лекций $\dot V>0$ при $\mathbf y\neq0$ — неустойчивость.

Частый первый кандидат:
$$
V=\sum_i a_i y_i^2,\qquad a_i>0.
$$

<details>
<summary><b>Пример: $\dot y_1=-y_1,\ \dot y_2=-y_2$</b></summary>
<div>

<p>Берём $V=y_1^2+y_2^2$.</p>
<div>$$
\dot V=-2y_1^2-2y_2^2<0
$$</div>
<p>при $(y_1,y_2)\neq(0,0)$. Нулевое решение асимптотически устойчиво.</p>

</div>
</details>

## 7.4. Матричное уравнение Ляпунова

Для
$$
\dot{\mathbf y}=A\mathbf y
$$
рассматриваем
$$
\boxed{
HA+A^*H=-C.
}
$$

Если
$$
C=C^*>0,\qquad H=H^*>0,
$$
то
$$
V(\mathbf y)=(H\mathbf y,\mathbf y)
$$
и
$$
\dot V=-(C\mathbf y,\mathbf y)<0.
$$

Если
$$
\operatorname{Re}\lambda_j(A)<0
\qquad\forall j,
$$
то для любого $C=C^*>0$ существует единственное $H=H^*>0$.

Интеграл Ляпунова:
$$
\boxed{
H=
\int_0^\infty
e^{tA^*}Ce^{tA}\,dt.
}
$$

Эквивалентность:
$$
\boxed{
\operatorname{Re}\lambda_j(A)<0
\iff
\forall C>0\ \exists!\ H>0:
HA+A^*H=-C.
}
$$

<details>
<summary><b>Пример из лекций</b></summary>
<div>

<p>Пусть</p>
<div>$$
A=
\begin{pmatrix}
-1&1\\
0&-1
\end{pmatrix},
\qquad C=E.
$$</div>
<p>Тогда</p>
<div>$$
\boxed{
H=
\begin{pmatrix}
1/2&1/4\\
1/4&3/4
\end{pmatrix}.
}
$$</div>
<p>$H>0$, поэтому нулевое решение асимптотически устойчиво.</p>

</div>
</details>

## 7.5. Устойчивость по первому приближению

Для автономной нелинейной системы
$$
\dot{\mathbf y}=\mathbf f(\mathbf y),\qquad \mathbf f(0)=0
$$
линеаризуем:
$$
\boxed{
\dot{\mathbf y}=A\mathbf y+o(\|\mathbf y\|),
\qquad
A=D\mathbf f(0).
}
$$

Тогда:
$$
\boxed{
\operatorname{Re}\lambda_j(A)<0\ \forall j
\Rightarrow
0\ \text{асимптотически устойчиво}.
}
$$

$$
\boxed{
\exists\,\lambda_k:\operatorname{Re}\lambda_k(A)>0
\Rightarrow
0\ \text{неустойчиво}.
}
$$

Если все $\operatorname{Re}\lambda_j\le0$, но есть $\operatorname{Re}\lambda=0$, линейного приближения недостаточно.

### Положение равновесия

Если
$$
\mathbf f(\mathbf c)=0,
$$
то $\mathbf y=\mathbf c$ — положение равновесия.

Полагаем
$$
\mathbf z=\mathbf y-\mathbf c.
$$

Матрица линеаризации:
$$
\boxed{
A=D\mathbf f(\mathbf c).
}
$$

Далее применяются те же спектральные критерии.


# 8. Фазовые портреты автономных систем

Автономная система:
$$
\dot{\mathbf y}=\mathbf f(\mathbf y).
$$

- **Фазовое пространство** — пространство переменных $(y_1,\ldots,y_n)$.
- **Фазовая траектория** — проекция интегральной кривой на фазовое пространство.
- **Фазовый портрет** — совокупность фазовых траекторий.


## 8.1. Основные свойства

Из автономности:
$$
\mathbf y(t)\ \text{решение}
\quad\Rightarrow\quad
\mathbf y(t+c)\ \text{тоже решение}.
$$

Из единственности задачи Коши:
$$
\boxed{
\text{две фазовые траектории либо не пересекаются, либо совпадают}.
}
$$

Для непродолжаемых решений возможны три типа фазовых траекторий:

1. точка — положение равновесия;
2. замкнутая кривая — цикл;
3. кривая без самопересечений.

> Если непостоянная траектория вернулась в уже пройденную точку, из единственности следует периодичность.


# 9. Уравнения с частными производными первого порядка

В этом курсе УЧП первого порядка связываются с первыми интегралами систем ОДУ.


## 9.1. Первые интегралы системы ОДУ

Пусть
$$
\dot{\mathbf y}=\mathbf f(t,\mathbf y).
$$

$\Phi(t,\mathbf y)$ называется первым интегралом, если вдоль любого решения
$$
\boxed{
\Phi(t,\mathbf y(t))=\mathrm{const}.
}
$$

Эквивалентный критерий:
$$
\boxed{
\frac{\partial\Phi}{\partial t}
+
\sum_{j=1}^n
f_j(t,\mathbf y)
\frac{\partial\Phi}{\partial y_j}
=0.
}
$$

Если $\Phi_1,\ldots,\Phi_k$ — первые интегралы, то $F(\Phi_1,\ldots,\Phi_k)$ тоже первый интеграл.

Функциональная независимость:
$$
\operatorname{rank}
\frac{\partial(\Phi_1,\ldots,\Phi_k)}
{\partial(t,y_1,\ldots,y_n)}
=k.
$$

Локально система из $n$ ОДУ имеет не более $n$ функционально независимых первых интегралов.

<details>
<summary><b>Пример из лекций</b></summary>
<div>

<p>Для системы</p>
<div>$$
x'=2x+y,\qquad y'=y
$$</div>
<p>первые интегралы:</p>
<div>$$
\Phi_1=ye^{-t},\qquad
\Phi_2=(x+y)e^{-2t}.
$$</div>
<p>Любой первый интеграл локально имеет вид</p>
<div>$$
F\!\left(ye^{-t},(x+y)e^{-2t}\right).
$$</div>

</div>
</details>

## 9.2. Общее решение линейного УЧП

$$
\frac{\partial u}{\partial t}
+
\sum_{j=1}^{n}
f_j(t,\mathbf y)
\frac{\partial u}{\partial y_j}
=0.
$$

Характеристическая система:
$$
\boxed{
\frac{dy_j}{dt}=f_j(t,\mathbf y).
}
$$

Вдоль характеристики:
$$
\boxed{
\frac{du}{dt}=0.
}
$$

Если $\Phi_1,\ldots,\Phi_n$ — функционально независимые первые интегралы характеристической системы:
$$
\boxed{
u(t,\mathbf y)=
F(\Phi_1,\ldots,\Phi_n).
}
$$

<details>
<summary><b>Пример: УЧП из системы выше</b></summary>
<div>

<div>$$
u_t+(2x+y)u_x+yu_y=0.
$$</div>
<p>Характеристики: $x'=2x+y$, $y'=y$.</p>
<div>$$
\boxed{
u(t,x,y)=
F\!\left(ye^{-t},(x+y)e^{-2t}\right).
}
$$</div>

</div>
</details>

## 9.3. Задача Коши для УЧП

Для
$$
u_t+\sum_{j=1}^{n}f_j(t,\mathbf y)u_{y_j}=0
$$
задаём
$$
\boxed{
u(t_0,\mathbf y)=\varphi(\mathbf y).
}
$$

Алгоритм:
1. решить характеристическую систему;
2. найти первые интегралы;
3. записать $u=F(\Phi_1,\ldots,\Phi_n)$;
4. подставить начальное условие и определить $F$.

<details>
<summary><b>Пример из лекций</b></summary>
<div>

<div>$$
u_t-u_y=0,\qquad u(0,y)=e^y.
$$</div>
<p>Характеристика:</p>
<div>$$
\frac{dy}{dt}=-1
\Rightarrow
y+t=C.
$$</div>
<p>$u=F(y+t)$, а из $F(y)=e^y$:</p>
<div>$$
\boxed{u(t,y)=e^{t+y}.}
$$</div>

</div>
</details>

## 9.4. Общая характеристическая система

Для
$$
f_0(t,\mathbf y)u_t+
\sum_{j=1}^{n}f_j(t,\mathbf y)u_{y_j}=0
$$
вводим параметр $s$:
$$
\boxed{
\frac{dt}{ds}=f_0,\qquad
\frac{dy_j}{ds}=f_j,\qquad
\frac{du}{ds}=0.
}
$$

То есть $u$ постоянно вдоль характеристик в пространстве $(t,\mathbf y)$.


## 9.5. Задача Коши на поверхности

Пусть
$$
S=\{(t,\mathbf y):g(t,\mathbf y)=0\},
\qquad
\nabla g\neq0,
$$
и
$$
u|_S=\psi.
$$

Для
$$
f_0u_t+\sum_{j=1}^{n}f_ju_{y_j}=0
$$
условие нехарактеристичности:
$$
\boxed{
\nabla g\cdot
(f_0,f_1,\ldots,f_n)\neq0.
}
$$

Если оно выполнено, задача Коши локально имеет единственное решение.

Если нарушено, возможны отсутствие решения или неединственность.

Для гиперплоскости $t=t_0$:
$$
\boxed{
f_0(t_0,\mathbf y)\neq0.
}
$$
