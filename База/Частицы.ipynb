# Карта частиц

Интерактивный справочник по частицам и связям между ними. Демонстрационная версия использует около 50 состояний из набора PDG 2024.

## [Открыть интерактивную карту частиц →](../particles/)

На карте доступны поиск по имени, символу и PDG ID, фильтры по семействам, включение типов взаимодействий, масштабирование и перемещение. Нажатие на частицу показывает её основные параметры и каналы распада.

> На телефоне карта поддерживает перетаскивание и pinch-to-zoom; фильтры открываются сбоку, а карточка частицы — снизу.
